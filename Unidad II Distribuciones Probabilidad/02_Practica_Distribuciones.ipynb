## Ejercicio 1: Distribución Binomial

### Contexto Económico - Cartera de Inversiones

Un banco comercial tiene una cartera de **80 préstamos empresariales**. Según datos históricos, la **probabilidad de default (incumplimiento) es del 8%** para este tipo de préstamos.

El departamento de riesgo necesita evaluar la probabilidad de pérdidas en la cartera.

### Preguntas:

1. ¿Cuál es la probabilidad de que **exactamente 5 préstamos** entren en default?
2. ¿Cuál es la probabilidad de que **al menos 10 préstamos** entren en default?
3. ¿Cuántos defaults se esperan en promedio?
4. ¿Cuál es el intervalo típico de defaults (usando ±1 desviación estándar)?
5. Graficar la distribución y simular 500 carteras.

**Aplicación:** Gestión de riesgo crediticio, provisionamiento de capital

### 📚 Conceptos Fundamentales: Distribución Binomial

**Definición**

La distribución Binomial modela el **número de éxitos en n ensayos independientes** de Bernoulli, cada uno con la misma probabilidad de éxito p.

**Notación:** X ~ Binomial(n, p)

**Función de Probabilidad (PMF)**

$$P(X = k) = \binom{n}{k} p^k (1-p)^{n-k}$$

Donde:
* n = número de ensayos
* k = número de éxitos (0, 1, 2, ..., n)
* p = probabilidad de éxito en cada ensayo
* $\binom{n}{k} = \frac{n!}{k!(n-k)!}$ = combinaciones de n en k

**Parámetros**
* **Esperanza:** E(X) = np
* **Varianza:** Var(X) = np(1 - p)
* **Desviación estándar:** σ(X) = √[np(1-p)]

**Supuestos importantes:**
1. Número fijo de ensayos (n)
2. Cada ensayo es independiente
3. Solo dos resultados posibles (éxito/fracaso)
4. Probabilidad constante (p) en todos los ensayos

**Cuándo usarla:**
* Número de clientes que compran de n visitantes
* Número de emails abiertos de n enviados
* Número de productos defectuosos en un lote
* Número de conversiones en una campaña

**Relación con otras distribuciones:**
* Si n = 1 → se reduce a Bernoulli
* Si n es grande y p pequeño (np ≈ constante) → se aproxima a Poisson

```python
# Parámetros del problema
n = 80  # número de préstamos en la cartera
p = 0.08  # probabilidad de default (incumplimiento)

# Crear distribución Binomial
X = stats.binom(n, p)

print("=" * 70)
print("DISTRIBUCIÓN BINOMIAL: Defaults en Cartera de Préstamos")
print("=" * 70)
print(f"Cartera: n = {n} préstamos empresariales")
print(f"Tasa histórica de default: p = {p*100}%")
print("=" * 70)

# 1. Probabilidad de exactamente 5 defaults
prob_5 = X.pmf(5)
print(f"\n1. P(X = 5 defaults) = {prob_5:.4f} ({prob_5*100:.2f}%)")
print(f"   Interpretación: En {prob_5*100:.2f}% de las carteras similares,")
print(f"   exactamente 5 préstamos entrarán en default")

# 2. Probabilidad de al menos 10 defaults (escenario de alto riesgo)
prob_10_o_mas = 1 - X.cdf(9)  # P(X >= 10)
porcentaje_cartera = (10/n)*100
print(f"\n2. P(X ≥ 10 defaults) = {prob_10_o_mas:.4f} ({prob_10_o_mas*100:.2f}%)")
print(f"   Interpretación: Probabilidad de escenario de ALTO RIESGO")
print(f"   (más del {porcentaje_cartera:.1f}% de la cartera en default)")
print(f"   Esto requeriría provisiones adicionales de capital")

# 3. Parámetros esperados (gestión de riesgo)
mu = X.mean()
std = X.std()
var = X.var()

print(f"\n3. PARÁMETROS ESPERADOS (Gestión de Riesgo):")
print(f"   E(X) = np = {mu:.2f} defaults esperados")
print(f"   Var(X) = np(1-p) = {var:.2f}")
print(f"   σ(X) = {std:.2f} defaults")
print(f"\n   💼 Interpretación para el CFO:")
print(f"   - En promedio: {mu:.1f} préstamos entrarán en default")
print(f"   - Variabilidad típica: ±{std:.1f} defaults")
print(f"   - Tasa esperada: {(mu/n)*100:.1f}% de la cartera")

# 4. Intervalo típico (±1 desviación estándar)
print(f"\n4. INTERVALO TÍPICO (±1σ):")
print(f"   [{mu - std:.1f}, {mu + std:.1f}] defaults")
print(f"   Esto cubre aproximadamente el 68% de los escenarios")
print(f"   Rango esperado: {(mu-std)/n*100:.1f}% - {(mu+std)/n*100:.1f}% de la cartera")

# 5. Probabilidades de escenarios clave para provisiones
print(f"\n5. ESCENARIOS PARA PROVISIONES DE CAPITAL:")
escenarios = {
    3: "Escenario optimista (3.75% de defaults)",
    6: "Escenario base ({:.1f}% - cerca del esperado)".format((6/n)*100),
    10: "Escenario pesimista (12.5% - alto riesgo)",
    15: "Escenario de crisis (18.75% - riesgo crítico)"
}

for k, descripcion in escenarios.items():
    prob = X.pmf(k)
    acum = X.cdf(k)
    print(f"   P(X = {k:2d}) = {prob:.4f} ({prob*100:.2f}%) - {descripcion}")
    print(f"            P(X ≤ {k:2d}) = {acum:.4f}")

# Visualización
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Gráfico 1: Función de probabilidad con zonas de riesgo
valores = np.arange(0, min(n+1, 25))  # Limitar rango para mejor visualización
probs = X.pmf(valores)

# Colorear según nivel de riesgo
colores = ['green' if v <= 5 else 'yellow' if v <= 9 else 'red' for v in valores]
ax1.bar(valores, probs, color=colores, alpha=0.7, edgecolor='black')

# Líneas de referencia
ax1.axvline(mu, color='blue', linestyle='--', linewidth=2, label=f'E(X) = {mu:.1f}')
ax1.axvspan(mu - std, mu + std, alpha=0.2, color='blue', label=f'±1σ [{mu-std:.1f}, {mu+std:.1f}]')

ax1.set_xlabel('Número de defaults en la cartera', fontsize=11)
ax1.set_ylabel('Probabilidad', fontsize=11)
ax1.set_title(f'Distribución Binomial(n={n}, p={p})\nVerde: Bajo riesgo | Amarillo: Moderado | Rojo: Alto', 
              fontsize=12, fontweight='bold')
ax1.legend(fontsize=10)
ax1.grid(axis='y', alpha=0.3)

# Gráfico 2: CDF con umbrales de decisión
valores_ext = np.arange(0, 21)
ax2.step(valores_ext, X.cdf(valores_ext), where='post', color='darkblue', linewidth=2, label='CDF')
ax2.fill_between(valores_ext, 0, X.cdf(valores_ext), step='post', alpha=0.2, color='blue')

# Umbrales de decisión
ax2.axhline(0.95, color='orange', linestyle='--', alpha=0.7, linewidth=1.5, label='95% confianza')
ax2.axvline(10, color='red', linestyle='--', alpha=0.7, linewidth=1.5, label='Umbral alto riesgo')

ax2.set_xlabel('Número de defaults', fontsize=11)
ax2.set_ylabel('Probabilidad acumulada P(X ≤ k)', fontsize=11)
ax2.set_title('Función de Distribución Acumulada (CDF)\nPara Análisis de Provisiones', 
              fontsize=12, fontweight='bold')
ax2.legend(fontsize=10)
ax2.grid(alpha=0.3)

plt.tight_layout()
plt.show()

# Simulación de 500 carteras
print(f"\n6. SIMULACIÓN DE 500 CARTERAS:")
np.random.seed(42)
simulacion = X.rvs(size=500)

media_sim = simulacion.mean()
std_sim = simulacion.std()
defaults_max = simulacion.max()
defaults_min = simulacion.min()

print(f"   Media de defaults: {media_sim:.2f} (teórico: {mu:.2f}) ✓")
print(f"   Desv. Est.: {std_sim:.2f} (teórico: {std:.2f}) ✓")
print(f"   Rango observado: [{defaults_min}, {defaults_max}] defaults")
print(f"   Carteras con ≥ 10 defaults: {np.sum(simulacion >= 10)} ({np.sum(simulacion >= 10)/500*100:.1f}%)")

print("\n" + "=" * 70)
print("💼 CONCLUSIÓN PARA GESTIÓN DE RIESGO:")
print("=" * 70)
print(f"• Provisionar capital para cubrir ~{mu + 2*std:.0f} defaults (percentil 95)")
print(f"• Monitorear alertas tempranas si defaults superan {mu + std:.0f}")
print(f"• Revisar política crediticia si defaults ≥ 10 (ocurre {prob_10_o_mas*100:.2f}% del tiempo)")
print("=" * 70)
```

## Ejercicio 2: Distribución Normal

### Contexto Económico - Salarios en el Sector Empresarial

El departamento de Recursos Humanos de una asociación empresarial realizó un estudio salarial en el sector de servicios profesionales. Los **salarios mensuales brutos** de analistas junior siguen una distribución normal con:

* **Media (μ)**: $85,000 pesos mensuales
* **Desviación estándar (σ)**: $12,000 pesos

### Preguntas:

1. ¿Qué proporción de analistas gana entre $75,000 y $95,000?
2. ¿Qué proporción gana más de $100,000 (considerado salario alto)?
3. Determinar el **intervalo del 95% central** de salarios
4. Verificar la **regla empírica** (68-95-99.7%)
5. ¿Cuál es el salario mínimo para estar en el **top 10%**?

**Aplicación:** Benchmarking salarial, negociación colectiva, presupuestación de nómina

### 📚 Conceptos Fundamentales: Distribución Normal

**Definición**

La distribución Normal (también llamada Gaussiana o campana de Gauss) es la distribución continua más importante en estadística. Modela variables que se concentran alrededor de un valor central.

**Notación:** X ~ N(μ, σ²)

**Función de Densidad (PDF)**

$$f(x) = \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{(x-\mu)^2}{2\sigma^2}}$$

Donde:
* μ (mu) = media (centro de la distribución)
* σ (sigma) = desviación estándar (dispersión)
* σ² = varianza

**Parámetros**
* **Esperanza:** E(X) = μ
* **Varianza:** Var(X) = σ²
* **Desviación estándar:** σ(X) = σ

**Propiedades importantes:**

1. **Simétrica:** La distribución es simétrica alrededor de μ
2. **Forma de campana:** Concentrada en el centro, colas que se extienden a ±∞
3. **Media = Mediana = Moda:** Todas iguales a μ
4. **Área total = 1:** Toda el área bajo la curva suma 1

**Regla Empírica (68-95-99.7):**

* **68%** de los datos están en [μ - σ, μ + σ]
* **95%** de los datos están en [μ - 2σ, μ + 2σ]
* **99.7%** de los datos están en [μ - 3σ, μ + 3σ]

**Distribución Normal Estándar:**

Cuando μ = 0 y σ = 1 → Z ~ N(0, 1)

**Estandarización (Z-score):**

$$Z = \frac{X - \mu}{\sigma}$$

Transforma cualquier normal a la estándar.

**Cuándo usarla:**
* Alturas de personas
* Pesos de productos
* Errores de medición
* Puntajes de pruebas
* Tiempos de respuesta
* Ventas diarias
* Temperaturas

**Por qué es tan importante:**
* Muchos fenómenos naturales siguen esta distribución
* **Teorema Central del Límite:** Las medias muestrales tienden a ser normales
* Base de muchos métodos estadísticos (intervalos de confianza, pruebas de hipótesis)

```python
# Parámetros del estudio salarial
mu = 85000  # salario promedio mensual (pesos)
sigma = 12000  # desviación estándar

# Crear distribución Normal
X = stats.norm(loc=mu, scale=sigma)

print("=" * 70)
print("DISTRIBUCIÓN NORMAL: Salarios de Analistas Junior")
print("=" * 70)
print(f"Sector: Servicios Profesionales")
print(f"Media (μ): ${mu:,} pesos/mes")
print(f"Desviación estándar (σ): ${sigma:,} pesos/mes")
print("=" * 70)

# 1. Probabilidad de ganar entre $75,000 y $95,000
salario_bajo = 75000
salario_alto = 95000
prob_entre = X.cdf(salario_alto) - X.cdf(salario_bajo)

print(f"\n1. P(${salario_bajo:,} ≤ X ≤ ${salario_alto:,}) = {prob_entre:.4f} ({prob_entre*100:.2f}%)")
print(f"   Interpretación: {prob_entre*100:.1f}% de los analistas ganan")
print(f"   en este rango (salarios típicos del mercado)")

# Cálculo con Z-scores
z1 = (salario_bajo - mu) / sigma
z2 = (salario_alto - mu) / sigma
print(f"\n   Z-scores: z1 = {z1:.2f}, z2 = {z2:.2f}")
print(f"   ({z1:.2f} y {z2:.2f} desviaciones estándar desde la media)")

# 2. Probabilidad de ganar más de $100,000 (salario alto)
salario_alto_umbral = 100000
prob_alto = 1 - X.cdf(salario_alto_umbral)

print(f"\n2. P(X > ${salario_alto_umbral:,}) = {prob_alto:.4f} ({prob_alto*100:.2f}%)")
print(f"   Interpretación: {prob_alto*100:.1f}% de los analistas ganan más de")
print(f"   ${salario_alto_umbral:,} (segmento de salarios altos)")

z_alto = (salario_alto_umbral - mu) / sigma
print(f"\n   Z-score: z = {z_alto:.2f}")
print(f"   ({abs(z_alto):.2f} desviaciones estándar por encima de la media)")

# 3. Intervalo del 95% central de salarios
# Queremos el intervalo [a, b] tal que P(a ≤ X ≤ b) = 0.95
# con 2.5% en cada cola
limite_inferior = X.ppf(0.025)  # percentil 2.5%
limite_superior = X.ppf(0.975)  # percentil 97.5%

print(f"\n3. INTERVALO DEL 95% CENTRAL:")
print(f"   [${limite_inferior:,.0f}, ${limite_superior:,.0f}]")
print(f"   Fórmula: μ ± 1.96σ = [${mu - 1.96*sigma:,.0f}, ${mu + 1.96*sigma:,.0f}]")
print(f"\n   💼 Interpretación para RRHH:")
print(f"   - 95% de los analistas ganan en este rango")
print(f"   - Usar como referencia para ofertas salariales competitivas")
print(f"   - Salarios fuera de este rango son atípicos (2.5% cada cola)")

# 4. Regla empírica (68-95-99.7) aplicada a salarios
print(f"\n4. REGLA EMPÍRICA (68-95-99.7):")

print(f"\n   ✓ 68% de los salarios están en μ ± 1σ:")
rango_1s_inf = mu - sigma
rango_1s_sup = mu + sigma
print(f"      [${rango_1s_inf:,.0f}, ${rango_1s_sup:,.0f}]")
prob_1sigma = X.cdf(rango_1s_sup) - X.cdf(rango_1s_inf)
print(f"      Probabilidad real: {prob_1sigma:.4f} ({prob_1sigma*100:.2f}%)")
print(f"      Rango salarial 'común'")

print(f"\n   ✓ 95% de los salarios están en μ ± 2σ:")
rango_2s_inf = mu - 2*sigma
rango_2s_sup = mu + 2*sigma
print(f"      [${rango_2s_inf:,.0f}, ${rango_2s_sup:,.0f}]")
prob_2sigma = X.cdf(rango_2s_sup) - X.cdf(rango_2s_inf)
print(f"      Probabilidad real: {prob_2sigma:.4f} ({prob_2sigma*100:.2f}%)")
print(f"      Rango para ofertas salariales competitivas")

print(f"\n   ✓ 99.7% de los salarios están en μ ± 3σ:")
rango_3s_inf = mu - 3*sigma
rango_3s_sup = mu + 3*sigma
print(f"      [${rango_3s_inf:,.0f}, ${rango_3s_sup:,.0f}]")
prob_3sigma = X.cdf(rango_3s_sup) - X.cdf(rango_3s_inf)
print(f"      Probabilidad real: {prob_3sigma:.4f} ({prob_3sigma*100:.2f}%)")
print(f"      Fuera de este rango: salarios extremos (outliers)")

# 5. Salario mínimo para estar en el top 10%
percentil_90 = X.ppf(0.90)
print(f"\n5. SALARIO MÍNIMO PARA TOP 10%:")
print(f"   ${percentil_90:,.0f} pesos/mes (percentil 90)")
print(f"   Los analistas con salario ≥ ${percentil_90:,.0f} están en el top 10%")
print(f"   Esto es {((percentil_90 - mu)/mu)*100:+.1f}% sobre el promedio")

# Visualización
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Rango de valores para graficar
x_vals = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y_vals = X.pdf(x_vals)

# Gráfico 1: Densidad con rango salarial típico
axes[0, 0].plot(x_vals, y_vals, 'b-', linewidth=2, label='Densidad')
axes[0, 0].fill_between(x_vals, 0, y_vals, where=(x_vals >= salario_bajo) & (x_vals <= salario_alto),
                         color='lightgreen', alpha=0.5, label=f'P(${salario_bajo:,} ≤ X ≤ ${salario_alto:,})')
axes[0, 0].axvline(mu, color='red', linestyle='--', linewidth=2, label=f'μ = ${mu:,}')
axes[0, 0].axvline(salario_alto_umbral, color='orange', linestyle=':', linewidth=2, label=f'Salario alto (${salario_alto_umbral:,})')
axes[0, 0].set_xlabel('Salario mensual (pesos)', fontsize=11)
axes[0, 0].set_ylabel('Densidad', fontsize=11)
axes[0, 0].set_title(f'Distribución Normal de Salarios\nμ=${mu:,}, σ=${sigma:,}', fontsize=12, fontweight='bold')
axes[0, 0].legend(fontsize=9)
axes[0, 0].grid(alpha=0.3)

# Gráfico 2: Regla empírica con rangos salariales
axes[0, 1].plot(x_vals, y_vals, 'b-', linewidth=2)
colores_regla = ['lightgreen', 'lightyellow', 'lightcoral']
labels_regla = ['±1σ (68%)', '±2σ (95%)', '±3σ (99.7%)']

for k in [1, 2, 3]:
    axes[0, 1].axvspan(mu - k*sigma, mu + k*sigma, alpha=0.2 + 0.1*k, 
                       color=colores_regla[k-1])
    axes[0, 1].axvline(mu - k*sigma, color='gray', linestyle='--', alpha=0.5, linewidth=1)
    axes[0, 1].axvline(mu + k*sigma, color='gray', linestyle='--', alpha=0.5, linewidth=1)
    
axes[0, 1].axvline(mu, color='red', linestyle='--', linewidth=2, label=f'Media = ${mu:,}')
axes[0, 1].set_xlabel('Salario mensual (pesos)', fontsize=11)
axes[0, 1].set_ylabel('Densidad', fontsize=11)
axes[0, 1].set_title('Regla Empírica: 68%-95%-99.7%\nRangos Salariales Estándar', fontsize=12, fontweight='bold')

# Etiquetas de porcentaje
axes[0, 1].text(mu, max(y_vals)*0.85, '68%\nSalarios\nComunes', ha='center', fontsize=10, 
                bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.5))
axes[0, 1].text(mu, max(y_vals)*0.6, '95%', ha='center', fontsize=11, fontweight='bold')
axes[0, 1].text(mu, max(y_vals)*0.35, '99.7%', ha='center', fontsize=10)
axes[0, 1].legend(fontsize=9)
axes[0, 1].grid(alpha=0.3)

# Gráfico 3: CDF con percentiles clave
axes[1, 0].plot(x_vals, X.cdf(x_vals), 'darkblue', linewidth=2, label='CDF')
axes[1, 0].axhline(0.5, color='red', linestyle='--', alpha=0.5, linewidth=1.5, label=f'Mediana (P50) = ${mu:,}')
axes[1, 0].axhline(0.90, color='orange', linestyle='--', alpha=0.5, linewidth=1.5, label=f'P90 (Top 10%) = ${percentil_90:,.0f}')
axes[1, 0].axvline(mu, color='red', linestyle='--', alpha=0.3)
axes[1, 0].axvline(percentil_90, color='orange', linestyle='--', alpha=0.3)

axes[1, 0].set_xlabel('Salario mensual (pesos)', fontsize=11)
axes[1, 0].set_ylabel('Probabilidad acumulada P(X ≤ x)', fontsize=11)
axes[1, 0].set_title('Función de Distribución Acumulada (CDF)\nPercentiles Salariales', fontsize=12, fontweight='bold')
axes[1, 0].legend(fontsize=9)
axes[1, 0].grid(alpha=0.3)

# Gráfico 4: Simulación de 1000 analistas
np.random.seed(42)
simulacion = X.rvs(size=1000)
media_sim = simulacion.mean()
std_sim = simulacion.std()

axes[1, 1].hist(simulacion, bins=35, density=True, color='skyblue', 
                alpha=0.7, edgecolor='black', label=f'Simulación (1000 analistas)')
axes[1, 1].plot(x_vals, y_vals, 'r-', linewidth=2, label='Teórico')
axes[1, 1].axvline(media_sim, color='green', linestyle='--', 
                   linewidth=2, label=f'Media sim = ${media_sim:,.0f}')
axes[1, 1].axvline(mu, color='red', linestyle=':', linewidth=2, alpha=0.7, label=f'Media teórica = ${mu:,}')

axes[1, 1].set_xlabel('Salario mensual (pesos)', fontsize=11)
axes[1, 1].set_ylabel('Densidad', fontsize=11)
axes[1, 1].set_title(f'Simulación vs Teórico\nValidación del Modelo', fontsize=12, fontweight='bold')
axes[1, 1].legend(fontsize=9)
axes[1, 1].grid(alpha=0.3)

print(f"\nVALIDACIÓN DE SIMULACIÓN (1000 analistas):")
print(f"   Media simulada: ${media_sim:,.0f} (teórica: ${mu:,}) ✓")
print(f"   Desv. Est. simulada: ${std_sim:,.0f} (teórica: ${sigma:,}) ✓")
print(f"   En rango [μ-σ, μ+σ]: {np.sum((simulacion >= rango_1s_inf) & (simulacion <= rango_1s_sup))/10:.1f}% (esperado: 68%)")

plt.tight_layout()
plt.show()

print("\n" + "=" * 70)
print("💼 CONCLUSIONES PARA RECURSOS HUMANOS Y ADMINISTRACIÓN:")
print("=" * 70)
print(f"• Salario promedio del mercado: ${mu:,}/mes")
print(f"• Rango competitivo (95% central): ${limite_inferior:,.0f} - ${limite_superior:,.0f}")
print(f"• Top 10% del mercado: ≥ ${percentil_90:,.0f}/mes")
print(f"• Salarios > ${salario_alto_umbral:,}: {prob_alto*100:.1f}% (segmento premium)")
print(f"• Usar estos datos para: negociaciones, presupuesto nómina, retención talento")
print("=" * 70)
```

## Ejercicio 3: Estandarización (Z-scores)

### Contexto Económico - Comparación de Rendimientos de Inversiones

Una consultora financiera evalúa el desempeño relativo de dos fondos de inversión:

**Fondo A (Conservador):**
* Rendimiento mensual promedio: μ_A = 2.5%
* Volatilidad (desv. est.): σ_A = 1.2%
* Mes actual: 3.8% de rendimiento

**Fondo B (Agresivo):**
* Rendimiento mensual promedio: μ_B = 4.0%
* Volatilidad (desv. est.): σ_B = 2.5%
* Mes actual: 6.0% de rendimiento

### Preguntas:

1. Calcular Z-scores para diferentes niveles de rendimiento del Fondo A
2. ¿Cuál fondo tuvo mejor desempeño **relativo** este mes?
3. ¿Qué rendimiento necesita el Fondo A para igualar el desempeño relativo del Fondo B?
4. Identificar rendimientos atípicos (|Z| > 2)

**Aplicación:** Análisis comparativo de inversiones, evaluación de desempeño, gestión de portafolios

### 📚 Conceptos Fundamentales: Estandarización y Z-scores

**¿Qué es la estandarización?**

La estandarización es el proceso de **transformar una variable con cualquier distribución normal N(μ, σ²) a la distribución normal estándar N(0, 1)**.

**Fórmula del Z-score:**

$$Z = \frac{X - \mu}{\sigma}$$

Donde:
* X = valor original en la distribución
* μ = media de la distribución original
* σ = desviación estándar de la distribución original
* Z = valor estandarizado (número de desviaciones estándar desde la media)

**Interpretación del Z-score:**

El Z-score nos dice **cuántas desviaciones estándar está un valor por encima o por debajo de la media**:

* **Z = 0** → El valor está exactamente en la media
* **Z = +1** → El valor está 1 desviación estándar por encima de la media
* **Z = -1** → El valor está 1 desviación estándar por debajo de la media
* **Z = +2** → El valor está 2 desviaciones estándar por encima de la media
* **Z = -2.5** → El valor está 2.5 desviaciones estándar por debajo de la media

**Valores típicos de Z:**

* **|Z| < 1**: Valor común (dentro del 68% central)
* **1 ≤ |Z| < 2**: Valor algo inusual (entre 68% y 95%)
* **2 ≤ |Z| < 3**: Valor inusual (entre 95% y 99.7%)
* **|Z| ≥ 3**: Valor muy inusual o atípico (outlier)

**Propiedades de la estandarización:**

1. **Preserva la forma:** La distribución estandarizada sigue siendo normal
2. **Preserva probabilidades:** P(X ≤ a) = P(Z ≤ (a-μ)/σ)
3. **Preserva el orden:** Si X₁ < X₂, entonces Z₁ < Z₂
4. **Media = 0, Varianza = 1:** La distribución estandarizada siempre tiene E(Z) = 0 y Var(Z) = 1

**Proceso inverso (des-estandarización):**

Si conoces un Z-score y quieres encontrar el valor original X:

$$X = \mu + Z \cdot \sigma$$

**¿Por qué es útil la estandarización?**

1. **Comparar valores de diferentes distribuciones:**
   - "¿Quién tuvo mejor desempeño relativo: un estudiante con 85/100 en un examen difícil (μ=70, σ=10) o uno con 90/100 en un examen fácil (μ=80, σ=5)?"
   - Calculas Z-scores para ambos y comparas

2. **Identificar valores atípicos (outliers):**
   - Regla práctica: |Z| > 3 sugiere un outlier

3. **Usar tablas estadísticas:**
   - Las tablas normales están estandarizadas (Z ~ N(0,1))
   - Conviertes cualquier normal a la estándar para usar las tablas

4. **Calcular probabilidades:**
   - Es más fácil trabajar con la distribución estándar
   - Todas las distribuciones normales se comportan igual tras estandarizar

**Ejemplo intuitivo:**

Si las alturas de hombres tienen μ = 175 cm y σ = 8 cm:
* Una persona de 183 cm tiene Z = (183-175)/8 = +1.0 → "1 desviación estándar más alto que el promedio"
* Una persona de 167 cm tiene Z = (167-175)/8 = -1.0 → "1 desviación estándar más bajo que el promedio"

Ambos están **igualmente lejos** de la media (en términos relativos), solo en direcciones opuestas.

**Relación con percentiles:**

* Z = -1.96 → percentil 2.5
* Z = -1.64 → percentil 5
* Z = 0 → percentil 50 (mediana)
* Z = +1.64 → percentil 95
* Z = +1.96 → percentil 97.5

```python
# Parámetros de los fondos
# Fondo A (Conservador)
mu_A = 2.5  # rendimiento promedio mensual (%)
sigma_A = 1.2  # volatilidad (%)
rendimiento_A = 3.8  # rendimiento del mes actual (%)

# Fondo B (Agresivo)
mu_B = 4.0  # rendimiento promedio mensual (%)
sigma_B = 2.5  # volatilidad (%)
rendimiento_B = 6.0  # rendimiento del mes actual (%)

# Crear distribuciones para los fondos
Fondo_A = stats.norm(loc=mu_A, scale=sigma_A)
Fondo_B = stats.norm(loc=mu_B, scale=sigma_B)
Z = stats.norm(loc=0, scale=1)  # Normal estándar

print("=" * 75)
print("ESTANDARIZACIÓN: Comparación de Rendimientos de Fondos de Inversión")
print("=" * 75)
print(f"\nFondo A (Conservador): μ={mu_A}%, σ={sigma_A}% | Rendimiento actual: {rendimiento_A}%")
print(f"Fondo B (Agresivo):    μ={mu_B}%, σ={sigma_B}% | Rendimiento actual: {rendimiento_B}%")
print("=" * 75)

# 1. Calcular Z-scores para diferentes rendimientos del Fondo A
rendimientos_eval = [1.0, 2.5, 4.0, 5.5]
print("\n1. Z-SCORES PARA DIFERENTES RENDIMIENTOS (Fondo A):")
print("-" * 75)

for r in rendimientos_eval:
    z = (r - mu_A) / sigma_A
    prob = Fondo_A.cdf(r)
    
    print(f"   Rendimiento {r:.1f}%: Z = {z:+.2f}, P(X ≤ {r:.1f}%) = {prob:.4f} ({prob*100:.1f}%)")
    
    # Interpretación
    if abs(z) < 0.5:
        categoria = "NORMAL (cerca del promedio)"
    elif abs(z) < 1.0:
        categoria = "Ligeramente inusual"
    elif abs(z) < 2.0:
        categoria = "Moderadamente inusual"
    else:
        categoria = "¡EXTREMO! (outlier)"
    
    if z == 0:
        print(f"              → Exactamente en la media - {categoria}")
    elif z > 0:
        print(f"              → {abs(z):.1f} σ por ENCIMA del promedio - {categoria}")
    else:
        print(f"              → {abs(z):.1f} σ por DEBAJO del promedio - {categoria}")
    print()

# 2. Comparación: ¿Cuál fondo tuvo mejor desempeño RELATIVO?
z_A = (rendimiento_A - mu_A) / sigma_A
z_B = (rendimiento_B - mu_B) / sigma_B

print("\n2. COMPARACIÓN DE DESEMPEÑO RELATIVO:")
print("-" * 75)
print(f"\n   Fondo A: Rendimiento {rendimiento_A}% con μ={mu_A}%, σ={sigma_A}%")
print(f"            Z_A = ({rendimiento_A} - {mu_A}) / {sigma_A} = {z_A:.2f}")
print(f"            Interpretación: {z_A:.2f} desviaciones estándar sobre su promedio")
print(f"            Percentil: {Fondo_A.cdf(rendimiento_A)*100:.1f}%")

print(f"\n   Fondo B: Rendimiento {rendimiento_B}% con μ={mu_B}%, σ={sigma_B}%")
print(f"            Z_B = ({rendimiento_B} - {mu_B}) / {sigma_B} = {z_B:.2f}")
print(f"            Interpretación: {z_B:.2f} desviaciones estándar sobre su promedio")
print(f"            Percentil: {Fondo_B.cdf(rendimiento_B)*100:.1f}%")

if z_A > z_B:
    mejor = "A"
    diferencia = z_A - z_B
else:
    mejor = "B"
    diferencia = z_B - z_A

print(f"\n   🏆 CONCLUSIÓN: El Fondo {mejor} tuvo MEJOR desempeño relativo")
print(f"      (Z-score más alto = mejor rendimiento ajustado por volatilidad)")
print(f"      Diferencia de Z-scores: {diferencia:.2f} σ")

# 3. ¿Qué rendimiento necesita Fondo A para igualar a Fondo B?
rendimiento_A_equivalente = mu_A + z_B * sigma_A

print(f"\n3. EQUIVALENCIA DE RENDIMIENTOS:")
print("-" * 75)
print(f"\n   El Fondo B tuvo Z = {z_B:.2f}")
print(f"   Para el Fondo A lograr el mismo desempeño relativo, necesita:")
print(f"\n   x = μ_A + Z_B · σ_A = {mu_A} + {z_B:.2f} · {sigma_A} = {rendimiento_A_equivalente:.2f}%")
print(f"\n   Interpretación: Un rendimiento de {rendimiento_A_equivalente:.2f}% en el Fondo A")
print(f"   es equivalente (en términos relativos) a {rendimiento_B}% en el Fondo B")

# 4. Identificar rendimientos atípicos (|Z| > 2)
print(f"\n4. IDENTIFICACIÓN DE RENDIMIENTOS ATÍPICOS (|Z| > 2):")
print("-" * 75)

# Umbrales para Fondo A
umbral_alto_A = mu_A + 2*sigma_A
umbral_bajo_A = mu_A - 2*sigma_A
print(f"\n   Fondo A: Rendimientos atípicos si X < {umbral_bajo_A:.1f}% o X > {umbral_alto_A:.1f}%")
print(f"            (fuera del rango μ ± 2σ)")

if rendimiento_A > umbral_alto_A:
    print(f"            ⚠️ El rendimiento actual ({rendimiento_A}%) ES ATÍPICO (muy alto)")
elif rendimiento_A < umbral_bajo_A:
    print(f"            ⚠️ El rendimiento actual ({rendimiento_A}%) ES ATÍPICO (muy bajo)")
else:
    print(f"            ✓ El rendimiento actual ({rendimiento_A}%) es normal")

# Umbrales para Fondo B
umbral_alto_B = mu_B + 2*sigma_B
umbral_bajo_B = mu_B - 2*sigma_B
print(f"\n   Fondo B: Rendimientos atípicos si X < {umbral_bajo_B:.1f}% o X > {umbral_alto_B:.1f}%")

if rendimiento_B > umbral_alto_B:
    print(f"            ⚠️ El rendimiento actual ({rendimiento_B}%) ES ATÍPICO (muy alto)")
elif rendimiento_B < umbral_bajo_B:
    print(f"            ⚠️ El rendimiento actual ({rendimiento_B}%) ES ATÍPICO (muy bajo)")
else:
    print(f"            ✓ El rendimiento actual ({rendimiento_B}%) es normal")

# Visualización comparativa
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Gráfico 1: Comparación de distribuciones de ambos fondos
x_vals_A = np.linspace(mu_A - 4*sigma_A, mu_A + 4*sigma_A, 500)
y_vals_A = Fondo_A.pdf(x_vals_A)

x_vals_B = np.linspace(mu_B - 4*sigma_B, mu_B + 4*sigma_B, 500)
y_vals_B = Fondo_B.pdf(x_vals_B)

ax1.plot(x_vals_A, y_vals_A, 'b-', linewidth=2.5, label=f'Fondo A: N(μ={mu_A}%, σ={sigma_A}%)')
ax1.fill_between(x_vals_A, 0, y_vals_A, alpha=0.2, color='blue')

ax1.plot(x_vals_B, y_vals_B, 'r-', linewidth=2.5, label=f'Fondo B: N(μ={mu_B}%, σ={sigma_B}%)')
ax1.fill_between(x_vals_B, 0, y_vals_B, alpha=0.2, color='red')

# Marcar rendimientos actuales
ax1.axvline(rendimiento_A, color='blue', linestyle='--', linewidth=2, alpha=0.7)
ax1.text(rendimiento_A, max(y_vals_A)*0.8, f'{rendimiento_A}%\nZ={z_A:.2f}', 
         ha='center', fontsize=10, bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.8))

ax1.axvline(rendimiento_B, color='red', linestyle='--', linewidth=2, alpha=0.7)
ax1.text(rendimiento_B, max(y_vals_B)*0.6, f'{rendimiento_B}%\nZ={z_B:.2f}', 
         ha='center', fontsize=10, bbox=dict(boxstyle='round', facecolor='lightcoral', alpha=0.8))

ax1.set_xlabel('Rendimiento mensual (%)', fontsize=12)
ax1.set_ylabel('Densidad', fontsize=12)
ax1.set_title('Distribuciones Originales de Ambos Fondos\nRendimientos Actuales Marcados', 
              fontsize=13, fontweight='bold')
ax1.legend(fontsize=10)
ax1.grid(alpha=0.3)

# Gráfico 2: Distribución estandarizada (Z-scores)
z_vals = np.linspace(-4, 4, 500)
z_y_vals = Z.pdf(z_vals)

ax2.plot(z_vals, z_y_vals, 'purple', linewidth=3, label='Z ~ N(0, 1)')
ax2.fill_between(z_vals, 0, z_y_vals, alpha=0.15, color='purple')

# Marcar Z-scores de los fondos
ax2.axvline(z_A, color='blue', linestyle='--', linewidth=2.5, alpha=0.8, label=f'Fondo A: Z={z_A:.2f}')
ax2.axvline(z_B, color='red', linestyle='--', linewidth=2.5, alpha=0.8, label=f'Fondo B: Z={z_B:.2f}')

# Sombrear áreas
ax2.fill_between(z_vals, 0, z_y_vals, where=(z_vals >= -1) & (z_vals <= 1),
                 color='green', alpha=0.2, label='Normal: ±1σ (68%)')
ax2.fill_between(z_vals, 0, z_y_vals, where=((z_vals >= -2) & (z_vals < -1)) | ((z_vals > 1) & (z_vals <= 2)),
                 color='yellow', alpha=0.2, label='Inusual: 1-2σ')
ax2.fill_between(z_vals, 0, z_y_vals, where=(z_vals < -2) | (z_vals > 2),
                 color='red', alpha=0.2, label='Atípico: >±2σ')

# Líneas de referencia
for z in [-2, -1, 0, 1, 2]:
    ax2.axvline(z, color='gray', linestyle=':', alpha=0.3, linewidth=1)

ax2.set_xlabel('Z-score (desviaciones estándar)', fontsize=12)
ax2.set_ylabel('Densidad', fontsize=12)
ax2.set_title('Distribución Estandarizada\nComparación de Desempeño Relativo', 
              fontsize=13, fontweight='bold')
ax2.legend(fontsize=9, loc='upper left')
ax2.grid(alpha=0.3)

# Anotación del ganador
ganador_z = max(z_A, z_B)
ax2.annotate(f'🏆 Mejor\ndesempeño\nrelativo', 
             xy=(ganador_z, Z.pdf(ganador_z)), 
             xytext=(ganador_z + 0.8, Z.pdf(ganador_z) + 0.05),
             fontsize=10, fontweight='bold',
             bbox=dict(boxstyle='round', facecolor='gold', alpha=0.7),
             arrowprops=dict(arrowstyle='->', lw=2, color='black'))

plt.tight_layout()
plt.show()

print("\n" + "=" * 75)
print("💼 CONCLUSIONES PARA ANÁLISIS DE INVERSIONES:")
print("=" * 75)
print(f"• La estandarización (Z-scores) permite comparar rendimientos en escalas diferentes")
print(f"• Un Z-score alto indica mejor desempeño relativo (ajustado por volatilidad)")
print(f"• Fondo {mejor}: Z = {max(z_A, z_B):.2f} → Mejor rendimiento considerando su riesgo")
print(f"• Un rendimiento con |Z| > 2 es atípico (solo ocurre ~5% del tiempo)")
print(f"• Uso: Comparar fondos, identificar outliers, evaluar gestores de cartera")
print("=" * 75)
```

## Ejercicio 4: Teorema Central del Límite (TCL)

### Contexto Económico - Estimación de Ingresos Promedio

Una consultora económica necesita estimar el **ingreso mensual promedio** de hogares en una región. La distribución de ingresos individuales NO es normal (tiene sesgo positivo - muchos ingresos bajos, pocos muy altos), pero el TCL garantiza que el **promedio muestral** sí se distribuirá normalmente si el tamaño de muestra es suficiente.

### Objetivo:

Demostrar que la distribución de **medias muestrales** tiende a ser normal, **independientemente de la distribución poblacional**, y aplicar esto a:

1. Estimación de intervalos de confianza para el ingreso promedio
2. Diseño de tamaños de muestra para estudios económicos
3. Validación de supuestos para pruebas estadísticas

**Aplicación:** Encuestas de ingreso, estudios de mercado, política pública, auditoría de muestras

### 📚 Conceptos Fundamentales: Teorema Central del Límite (TCL)

**Enunciado del Teorema**

El Teorema Central del Límite (TCL) es uno de los teoremas más importantes de la estadística. Establece que:

> **Si tomamos muestras aleatorias de tamaño n de CUALQUIER población (con media μ y desviación estándar σ), la distribución de las medias muestrales se aproxima a una distribución normal a medida que n aumenta.**

**Formalmente:**

Si X₁, X₂, ..., Xₙ son variables aleatorias independientes con media μ y varianza σ², entonces:

$$\bar{X} = \frac{X_1 + X_2 + ... + X_n}{n}$$

Tiene distribución aproximadamente:

$$\bar{X} \sim N\left(\mu, \frac{\sigma^2}{n}\right)$$

O equivalentemente:

$$\bar{X} \sim N\left(\mu, \frac{\sigma}{\sqrt{n}}\right)^2$$

**Error Estándar de la Media:**

$$SE = \frac{\sigma}{\sqrt{n}}$$

* Mide la variabilidad de las medias muestrales
* Disminuye con √n (no con n)
* Para reducir el error a la mitad, necesitas 4 veces más muestras

**Aspectos clave:**

1. **NO importa la forma de la población original:**
   - Puede ser exponencial, uniforme, sesgada, bimodal, etc.
   - ¡Las medias muestrales SIEMPRE tienden a ser normales!

2. **¿Qué tan grande debe ser n?**
   - Regla general: **n ≥ 30** es suficiente en la mayoría de casos
   - Si la población ya es aproximadamente normal, n puede ser menor
   - Si la población es muy sesgada, necesitas n más grande

3. **La media de las medias = media poblacional:**
   - E(̄X) = μ

4. **La varianza disminuye con n:**
   - Var(̄X) = σ²/n

**Por qué es tan importante:**

* Justifica el uso de métodos basados en la normal incluso cuando la población no es normal
* Base de intervalos de confianza y pruebas de hipótesis
* Explica por qué las medias son más estables que valores individuales
* Permite hacer inferencia estadística con muestras

**Ejemplo intuitivo:**

Si lanzas un dado (NO normal, es uniforme), un resultado individual puede ser cualquier valor de 1 a 6 con igual probabilidad. Pero si lanzas 30 dados muchas veces y calculas el promedio cada vez, esos promedios seguirán una distribución normal centrada en 3.5.

```python
print("=" * 75)
print("TEOREMA CENTRAL DEL LÍMITE: Estimación de Ingresos Promedio")
print("=" * 75)

# Crear una población de ingresos con distribución NO NORMAL (Gamma sesgada)
# Simula ingresos reales: muchos bajos, pocos muy altos
np.random.seed(42)
shape_param = 2  # Parámetro de forma (sesgo)
scale_param = 25000  # Escala (en pesos)
poblacion = np.random.gamma(shape=shape_param, scale=scale_param, size=10000)

media_pob = poblacion.mean()
std_pob = poblacion.std()
mediana_pob = np.median(poblacion)

print(f"\nPOBLACIÓN DE INGRESOS (Distribución Gamma - NO NORMAL, sesgada):")
print(f"  Media poblacional: ${media_pob:,.0f}/mes")
print(f"  Mediana poblacional: ${mediana_pob:,.0f}/mes (< media por sesgo positivo)")
print(f"  Desv. Est. poblacional: ${std_pob:,.0f}")
print(f"  Tamaño población: {len(poblacion):,} hogares")
print(f"  Sesgo: Positivo (cola derecha larga - pocos ingresos muy altos)")

# Tomar muchas muestras de diferentes tamaños (simular encuestas repetidas)
tamaños_muestra = [5, 10, 30, 50]  # Diferentes tamaños de muestra
num_muestras = 1000  # Número de encuestas simuladas

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

for idx, n in enumerate(tamaños_muestra):
    # Tomar num_muestras muestras de tamaño n
    medias_muestrales = []
    for _ in range(num_muestras):
        muestra = np.random.choice(poblacion, size=n)
        medias_muestrales.append(muestra.mean())
    
    medias_muestrales = np.array(medias_muestrales)
    
    # Calcular estadísticas
    media_de_medias = medias_muestrales.mean()
    std_de_medias = medias_muestrales.std()
    
    # Error estándar teórico
    error_estandar_teorico = poblacion.std() / np.sqrt(n)
    
    print(f"\n{'='*70}")
    print(f"MUESTRAS DE TAMAÑO n = {n} (encuestas de {n} hogares):")
    print(f"{'='*70}")
    print(f"  Media de las {num_muestras} medias muestrales: ${media_de_medias:,.0f}")
    print(f"  Desv. Est. de las medias (observada): ${std_de_medias:,.0f}")
    print(f"  Error estándar teórico (SE = σ/√n): ${error_estandar_teorico:,.0f}")
    print(f"  Reducción de variabilidad vs población: {std_pob/std_de_medias:.1f}x")
    
    # Graficar
    ax = axes[idx]
    
    # Histograma de medias muestrales (ingresos promedio de encuestas)
    ax.hist(medias_muestrales, bins=35, density=True, 
            color='lightgreen', alpha=0.7, edgecolor='black', label=f'Medias de {num_muestras} encuestas')
    
    # Curva normal teórica predicha por el TCL
    x_vals = np.linspace(medias_muestrales.min(), medias_muestrales.max(), 100)
    normal_teorica = stats.norm.pdf(x_vals, loc=media_pob, scale=error_estandar_teorico)
    ax.plot(x_vals, normal_teorica, 'r-', linewidth=2.5, label='Normal teórica (TCL)')
    
    ax.axvline(media_pob, color='blue', linestyle='--', 
               linewidth=2, label=f'μ poblacional = ${media_pob:,.0f}')
    
    ax.set_xlabel('Ingreso promedio muestral ($/mes)', fontsize=10)
    ax.set_ylabel('Densidad', fontsize=10)
    ax.set_title(f'n = {n} hogares por encuesta\nSE = ${error_estandar_teorico:,.0f}', 
                 fontsize=11, fontweight='bold')
    ax.legend(fontsize=8, loc='upper right')
    ax.grid(alpha=0.3)
    
    # Anotación sobre normalidad
    if n >= 30:
        ax.text(0.5, 0.95, '✓ Aprox. Normal', transform=ax.transAxes,
                ha='center', va='top', fontsize=10, fontweight='bold',
                bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.7))
    else:
        ax.text(0.5, 0.95, 'Aún no muy normal', transform=ax.transAxes,
                ha='center', va='top', fontsize=9,
                bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.7))

plt.tight_layout()
plt.show()

print("\n" + "=" * 75)
print("CONCLUSIONES DEL TCL APLICADAS A ENCUESTAS DE INGRESOS:")
print("=" * 75)
print("\n1. 📊 A medida que n aumenta, la distribución de medias muestrales")
print("   se aproxima MÁS a una distribución normal (incluso con población sesgada)")
print(f"\n2. 🎯 La media de las medias muestrales ≈ media poblacional")
print(f"   (todas convergen a μ = ${media_pob:,.0f})")
print(f"\n3. 📉 El error estándar (SE) disminuye con √n:")
for n in tamaños_muestra:
    se = std_pob / np.sqrt(n)
    print(f"   n={n:2d}: SE = ${se:,.0f} (para duplicar precisión, necesitas 4x muestras)")
print(f"\n4. ✅ ¡Funciona incluso con población NO NORMAL (sesgada)!")
print(f"   La población de ingresos es Gamma (sesgada), pero las medias")
print(f"   muestrales sí son normales para n ≥ 30")

# Comparar distribución poblacional vs distribución de medias
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Población original de ingresos (sesgada, NO normal)
ax1.hist(poblacion, bins=60, density=True, color='salmon', 
         alpha=0.7, edgecolor='black', label='Población real')
ax1.axvline(media_pob, color='red', linestyle='--', linewidth=2.5, label=f'Media = ${media_pob:,.0f}')
ax1.axvline(mediana_pob, color='orange', linestyle=':', linewidth=2.5, label=f'Mediana = ${mediana_pob:,.0f}')

ax1.set_xlabel('Ingreso mensual ($/mes)', fontsize=12)
ax1.set_ylabel('Densidad', fontsize=12)
ax1.set_title('POBLACIÓN DE INGRESOS\n(Distribución Gamma - Sesgada, NO Normal)', 
              fontsize=13, fontweight='bold')
ax1.legend(fontsize=10)
ax1.grid(alpha=0.3)
ax1.text(0.98, 0.95, 'SESGO POSITIVO\n(cola derecha)', 
         transform=ax1.transAxes, ha='right', va='top',
         fontsize=10, bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.7))

# Distribución de medias muestrales (n=30 - tamaño típico de encuesta)
n = 30
num_encuestas = 1000
medias_finales = []
for _ in range(num_encuestas):
    muestra = np.random.choice(poblacion, size=n)
    medias_finales.append(muestra.mean())

medias_finales = np.array(medias_finales)
media_de_medias_final = medias_finales.mean()
std_de_medias_final = medias_finales.std()

ax2.hist(medias_finales, bins=40, density=True, color='lightgreen', 
         alpha=0.7, edgecolor='black', label=f'{num_encuestas} medias muestrales (n={n})')

# Curva normal teórica (predicha por TCL)
x_vals = np.linspace(min(medias_finales), max(medias_finales), 100)
se_teorico = std_pob / np.sqrt(n)
y_vals = stats.norm.pdf(x_vals, loc=media_pob, scale=se_teorico)
ax2.plot(x_vals, y_vals, 'r-', linewidth=3, label='Normal teórica (TCL)')

ax2.axvline(media_de_medias_final, color='green', linestyle='--', linewidth=2.5, 
            label=f'Media de medias = ${media_de_medias_final:,.0f}')

ax2.set_xlabel('Ingreso promedio muestral ($/mes)', fontsize=12)
ax2.set_ylabel('Densidad', fontsize=12)
ax2.set_title(f'DISTRIBUCIÓN DE MEDIAS MUESTRALES\n(n={n} hogares, {num_encuestas} encuestas)\n¡APROXIMADAMENTE NORMAL!', 
              fontsize=13, fontweight='bold')
ax2.legend(fontsize=9)
ax2.grid(alpha=0.3)
ax2.text(0.98, 0.95, '✓ NORMAL\n(por TCL)', 
         transform=ax2.transAxes, ha='right', va='top',
         fontsize=11, fontweight='bold',
         bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.8))

plt.tight_layout()
plt.show()

print(f"\nVERIFICACIÓN NUMÉRICA (n={n}):")
print(f"  Media de medias: ${media_de_medias_final:,.0f} (esperado: ${media_pob:,.0f}) ✓")
print(f"  Desv. Est. de medias: ${std_de_medias_final:,.0f} (esperado: ${se_teorico:,.0f}) ✓")
print(f"  Reducción de variabilidad: {std_pob/std_de_medias_final:.1f}x vs población")

print("\n" + "=" * 75)
print("💼 APLICACIONES PRÁCTICAS EN ECONOMÍA Y ADMINISTRACIÓN:")
print("=" * 75)
print(f"\n1. 📊 ENCUESTAS DE INGRESO:")
print(f"   Con n=30 hogares, el ingreso promedio muestral se distribuye ~N(μ, SE)")
print(f"   SE = ${se_teorico:,.0f} → Intervalo de confianza 95%: ±${1.96*se_teorico:,.0f}")

print(f"\n2. 🎯 DISEÑO DE MUESTRAS:")
print(f"   Para SE = $5,000: necesitas n = ({std_pob:,.0f}/5,000)² = {(std_pob/5000)**2:.0f} hogares")

print(f"\n3. ⚖️ JUSTIFICACIÓN DE MÉTODOS:")
print(f"   Intervalos de confianza y pruebas t requieren normalidad de la media")
print(f"   El TCL garantiza esto para n ≥ 30, incluso con población sesgada")

print(f"\n4. 📈 AUDITORÍA Y MUESTREO:")
print(f"   El promedio de una muestra aleatoria de facturas/transacciones")
print(f"   es un estimador confiable del promedio poblacional")

print("\n" + "=" * 75)
print("✓ Teorema Central del Límite verificado experimentalmente")
print("=" * 75)
```

## Resumen y Conclusiones

### Distribuciones y Conceptos Cubiertos (Aplicados a Ciencias Económicas)

| Concepto | Tipo | Uso en Economía/Administración | scipy.stats |
|----------|------|----------------------------------|-------------|
| **Binomial** | Discreta | Defaults de crédito, éxito/fracaso en inversiones, control de calidad | `binom(n, p)` |
| **Normal** | Continua | Salarios, precios, ventas, rendimientos financieros, errores de pronóstico | `norm(μ, σ)` |
| **Z-scores** | Transformación | Comparar indicadores en diferentes escalas, evaluar desempeño relativo | `(x-μ)/σ` |
| **TCL** | Teorema | Estimación de medias poblacionales, diseño de encuestas, intervalos de confianza | Muestreo |

### Métodos Esenciales en scipy.stats

**Para distribuciones discretas (Binomial):**
* **`.pmf(k)`**: Probabilidad exacta - P(X = k)
* **`.cdf(x)`**: Probabilidad acumulada - P(X ≤ x)
* **`.mean()`, `.var()`, `.std()`**: Parámetros

**Para distribuciones continuas (Normal):**
* **`.pdf(x)`**: Función de densidad - f(x)
* **`.cdf(x)`**: Probabilidad acumulada - P(X ≤ x)
* **`.ppf(q)`**: Percentil q (inversa de CDF) - valor x tal que P(X ≤ x) = q
* **`.rvs(size)`**: Generar muestras aleatorias

**Fórmulas clave:**
* **Z-score**: Z = (X - μ) / σ  (estandarización)
* **Error Estándar**: SE = σ / √n  (TCL)
* **Intervalo de Confianza 95%**: [μ - 1.96·SE, μ + 1.96·SE]

### Aplicaciones en Ciencias Económicas, Contables y Administración

#### 1. Distribución Binomial
* **Gestión de Riesgo Crediticio**: Estimar defaults en carteras de préstamos
* **Control de Calidad**: Productos defectuosos en lotes
* **Inversiones**: Probabilidad de éxito en proyectos independientes
* **Parámetros**: n (ensayos), p (probabilidad de éxito)
* **Ejemplo real**: 80 préstamos con 8% de default → Provisionar capital

#### 2. Distribución Normal
* **Recursos Humanos**: Benchmarking salarial, políticas de compensación
* **Finanzas**: Rendimientos de activos, precios de commodities
* **Ventas/Marketing**: Forecasting de demanda, presupuestación
* **Parámetros**: μ (media), σ (desviación estándar)
* **Regla 68-95-99.7**: El 95% de datos están en μ ± 2σ
* **Ejemplo real**: Salarios N(μ=$85k, σ=$12k) → Intervalo competitivo

#### 3. Estandarización (Z-scores)
* **Comparación de Inversiones**: Evaluar fondos con diferentes volatilidades
* **Evaluación de Desempeño**: Comparar empleados/sucursales en métricas distintas
* **Detección de Anomalías**: Identificar transacciones atípicas (|Z| > 2)
* **Fórmula**: Z = (X - μ) / σ
* **Interpretación**: Z indica cuántas σ está un valor de la media
* **Ejemplo real**: Comparar rendimiento 3.8% (Fondo A) vs 6% (Fondo B)

#### 4. Teorema Central del Límite (TCL)
* **Encuestas de Ingresos/Gastos**: Estimar promedios poblacionales con muestras
* **Auditoría**: Muestreo de facturas/transacciones para inferir totales
* **Estudios de Mercado**: Diseño de tamaño de muestra para encuestas
* **Intervalo de Confianza 95%**: [μ - 1.96·SE, μ + 1.96·SE] donde SE = σ/√n
* **Regla clave**: Para n ≥ 30, la media muestral es ~Normal, ¡incluso si la población NO lo es!
* **Ejemplo real**: Estimar ingreso promedio con encuesta de 30 hogares

### Conceptos Clave para Ciencias Económicas

1. **Parámetros vs Estadísticos**: 
   - Parámetros (μ, σ) describen la población (desconocidos)
   - Estadísticos (x̄, s) describen la muestra (calculables)

2. **Distribución de Probabilidad vs Distribución Muestral**:
   - Distribución de prob.: Comportamiento de la variable (ej: ingresos individuales)
   - Distribución muestral: Comportamiento del estadístico (ej: ingreso promedio de muestras)

3. **Inferencia Estadística**:
   - Usar muestras (datos observados) para hacer conclusiones sobre la población
   - El TCL justifica usar métodos basados en la Normal para medias muestrales

4. **Toma de Decisiones bajo Incertidumbre**:
   - Binomial: ¿Qué tan probable es un escenario de riesgo?
   - Normal: ¿Está este valor dentro del rango esperado?
   - Z-scores: ¿Cómo se compara este resultado con el estándar?
   - TCL: ¿Qué tan precisa es nuestra estimación del promedio?
