Librerías necesarias

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
```

En una materia de programación, se registró la cantidad de ejercicios resueltos correctamente por 12 estudiantes en una práctica:

2, 3, 1, 4, 2, 3, 2, 5, 4, 3, 2, 1


Datos

```python
datos = [2,3,1,4,2,3,2,5,4,3,2,1]
```

Ordenar datos

```python
datos_ordenados = sorted(datos)
print("Datos ordenados:", datos_ordenados)
```

Media

```python
media = np.mean(datos)
print("Media:", media)
```

Mediana

```python
mediana = np.median(datos)
print("Mediana:", mediana)
```

Moda

```python
moda = Counter(datos).most_common(1)[0][0]
print("Moda:", moda)
```

Varianza

```python
varianza = np.var(datos)
print("Varianza:", varianza)
```

Desvío estándar

```python
desvio = np.std(datos)
print("Desvío estándar:", desvio)
```

Rango

```python
rango = max(datos) - min(datos)
print("Rango:", rango)
```

Tabla de frecuencias

```python
tabla = pd.Series(datos).value_counts().sort_index()
print("\nTabla de frecuencias:")
print(tabla)
```

Gráfico de barras

```python
plt.bar(tabla.index, tabla.values)
plt.xlabel("Cantidad de ejercicios resueltos")
plt.ylabel("Frecuencia")
plt.title("Distribución de ejercicios resueltos")
plt.show()
```
