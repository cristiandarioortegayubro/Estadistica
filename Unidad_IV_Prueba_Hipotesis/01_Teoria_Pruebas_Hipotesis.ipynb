# Unidad IV: Prueba de Hipótesis, Regresión y ANOVA - Teoría

## Introducción

Las **pruebas de hipótesis** son procedimientos estadísticos que nos permiten tomar decisiones basadas en evidencia muestral. En negocios, son fundamentales para:

* Evaluar afirmaciones sobre parámetros poblacionales
* Comparar grupos (productos, mercados, estrategias)
* Modelar relaciones entre variables (regresión)
* Analizar efectos de múltiples factores (ANOVA)
* Tomar decisiones informadas con riesgo controlado

## Objetivos de aprendizaje

* Formular y contrastar hipótesis estadísticas
* Comprender errores Tipo I y Tipo II
* Realizar pruebas de hipótesis para medias y proporciones
* Aplicar regresión lineal simple
* Interpretar correlaciones
* Realizar análisis de varianza (ANOVA)
* Utilizar Python para análisis inferencial

## 1. Conceptos Fundamentales de Prueba de Hipótesis

### Hipótesis Científica vs Estadística

**Hipótesis científica**: Afirmación teórica sobre un fenómeno que puede ser investigada.

**Hipótesis estadística**: Afirmación específica sobre uno o más parámetros poblacionales que puede ser probada con datos.

### Hipótesis Nula (H₀) y Alternativa (H₁)

**Hipótesis Nula (H₀)**:
* Hipótesis de "no efecto", "no diferencia", o "status quo"
* Lo que asumimos verdadero hasta tener evidencia en contra
* Contiene igualdad (=, ≤, ≥)

**Hipótesis Alternativa (H₁ o Hₐ)**:
* Lo que queremos probar o demostrar
* Contradice a H₀
* Contiene desigualdad (≠, >, <)

### Tipos de Pruebas

#### Prueba Bilateral (dos colas)
```
H₀: μ = μ₀
H₁: μ ≠ μ₀
```
**Uso**: Detectar diferencia en cualquier dirección

#### Prueba Unilateral Derecha
```
H₀: μ ≤ μ₀
H₁: μ > μ₀
```
**Uso**: Probar si el parámetro es mayor

#### Prueba Unilateral Izquierda
```
H₀: μ ≥ μ₀
H₁: μ < μ₀
```
**Uso**: Probar si el parámetro es menor

### Ejemplos en Negocios

**Bilateral**: ¿El tiempo promedio de entrega es diferente de 2 días?
* H₀: μ = 2
* H₁: μ ≠ 2

**Unilateral derecha**: ¿La nueva campaña aumentó las ventas?
* H₀: μ ≤ μ₀ (ventas no aumentaron)
* H₁: μ > μ₀ (ventas aumentaron)

**Unilateral izquierda**: ¿El nuevo proceso redujo defectos?
* H₀: μ ≥ μ₀ (defectos no disminuyeron)
* H₁: μ < μ₀ (defectos disminuyeron)

## 2. Errores en Pruebas de Hipótesis

### Tabla de Decisiones

|                    | **H₀ es verdadera** | **H₀ es falsa** |
|--------------------|-------------------|------------------|
| **No rechazar H₀** | ✓ Decisión correcta (1-α) | **Error Tipo II** (β) |
| **Rechazar H₀**    | **Error Tipo I** (α) | ✓ Decisión correcta (Potencia = 1-β) |

### Error Tipo I (α)

**Definición**: Rechazar H₀ cuando en realidad es verdadera (falso positivo).

**Ejemplo**: Concluir que un tratamiento funciona cuando en realidad no tiene efecto.

**Probabilidad**: α = P(Rechazar H₀ | H₀ es verdadera)

**Nivel de significancia**: Típicamente α = 0.05 (5%), 0.01 (1%), o 0.10 (10%)

### Error Tipo II (β)

**Definición**: No rechazar H₀ cuando en realidad es falsa (falso negativo).

**Ejemplo**: No detectar que un tratamiento funciona cuando sí tiene efecto.

**Probabilidad**: β = P(No rechazar H₀ | H₀ es falsa)

### Potencia de la Prueba

**Definición**: Probabilidad de rechazar H₀ cuando es falsa (detectar un efecto real).

**Potencia** = 1 - β

**Factores que aumentan la potencia:**
* Mayor tamaño de muestra (n)
* Mayor nivel de significancia α
* Mayor tamaño del efecto (diferencia real más grande)
* Menor variabilidad poblacional

### Relación entre α y β

* Al disminuir α (ser más estrictos), aumenta β
* Al aumentar α (ser más permisivos), disminuye β
* **Trade-off**: Hay que balancear ambos errores según el contexto

### Importancia del Contexto

**Ejemplo médico**: Preferimos minimizar Error Tipo II (no detectar una enfermedad grave)

**Ejemplo legal**: Preferimos minimizar Error Tipo I (condenar a un inocente)

## 3. Pasos de una Prueba de Hipótesis

### Metodología Estándar

**Paso 1: Plantear las hipótesis**
* Definir H₀ y H₁ claramente
* Especificar el parámetro de interés

**Paso 2: Elegir nivel de significancia**
* Típicamente α = 0.05
* Ajustar según consecuencias de errores

**Paso 3: Calcular el estadístico de prueba**
* Depende del tipo de hipótesis (Z, t, F, etc.)
* Usar datos muestrales

**Paso 4: Determinar valor crítico o p-valor**
* **Método del valor crítico**: Comparar estadístico con valor de tabla
* **Método del p-valor**: Calcular probabilidad asociada al estadístico

**Paso 5: Tomar decisión**
* Rechazar o no rechazar H₀

**Paso 6: Conclusión en contexto**
* Interpretar el resultado en términos del problema original

### Método del Valor Crítico

**Regla de decisión:**
* Si |Estadístico| > Valor crítico → Rechazar H₀
* Si |Estadístico| ≤ Valor crítico → No rechazar H₀

### Método del p-valor

**Definición del p-valor**: Probabilidad de obtener un estadístico tan extremo o más que el observado, asumiendo que H₀ es verdadera.

**Regla de decisión:**
* Si p-valor ≤ α → Rechazar H₀ (evidencia significativa)
* Si p-valor > α → No rechazar H₀ (evidencia insuficiente)

**Interpretación:**
* p < 0.01: Evidencia muy fuerte contra H₀
* p < 0.05: Evidencia fuerte contra H₀
* p < 0.10: Evidencia moderada contra H₀
* p > 0.10: Evidencia débil o nula contra H₀

⚠️ **Importante**: "No rechazar H₀" NO significa "aceptar H₀" o "probar que H₀ es verdadera". Solo significa que no hay suficiente evidencia para rechazarla.

## 4. Pruebas de Hipótesis para la Media

### Prueba Z (cuando σ es conocida)

**Condiciones:**
* σ conocida
* Población normal o n ≥ 30

**Estadístico:**
$$Z = \frac{\bar{x} - \mu_0}{\sigma/\sqrt{n}}$$

Donde:
* x̄ = media muestral
* μ₀ = valor hipotético de la media
* σ = desviación estándar poblacional
* n = tamaño de muestra

**Valores críticos (de tablas Z):**
* Bilateral (α = 0.05): ±1.96
* Unilateral (α = 0.05): 1.645 o -1.645

### Prueba t (cuando σ es desconocida)

**Condiciones:**
* σ desconocida
* Población aproximadamente normal (especialmente si n < 30)

**Estadístico:**
$$t = \frac{\bar{x} - \mu_0}{s/\sqrt{n}}$$

Donde:
* s = desviación estándar muestral
* Grados de libertad: df = n - 1

**Valores críticos**: De tabla t con (n-1) gl

### Ejemplo Completo

Una empresa afirma que el tiempo promedio de entrega es 3 días. Una muestra de 36 entregas tiene x̄ = 3.4 días y s = 0.8 días. ¿Hay evidencia de que el tiempo real es diferente? (α = 0.05)

**Paso 1: Hipótesis**
* H₀: μ = 3
* H₁: μ ≠ 3 (bilateral)

**Paso 2: Nivel de significancia**
* α = 0.05

**Paso 3: Estadístico**
$$t = \frac{3.4 - 3}{0.8/\sqrt{36}} = \frac{0.4}{0.133} = 3.00$$

**Paso 4: Valor crítico**
* t_{0.025, 35} ≈ 2.03 (de tabla t)
* Como es bilateral, región crítica: t < -2.03 o t > 2.03

**Paso 5: Decisión**
* 3.00 > 2.03 → **Rechazar H₀**

**Paso 6: Conclusión**
* Con 95% de confianza, hay evidencia suficiente para concluir que el tiempo promedio de entrega es diferente de 3 días.

## 5. Prueba de Hipótesis para una Proporción

### Prueba Z para Proporción

**Condiciones:**
* np₀ ≥ 5 y n(1 - p₀) ≥ 5
* Muestra aleatoria

**Hipótesis:**
```
H₀: p = p₀
H₁: p ≠ p₀  (o p > p₀, o p < p₀)
```

**Estadístico:**
$$Z = \frac{\hat{p} - p_0}{\sqrt{\frac{p_0(1-p_0)}{n}}}$$

Donde:
* p̂ = x/n (proporción muestral)
* p₀ = proporción hipotética
* x = número de éxitos
* n = tamaño de muestra

### Ejemplo

Una empresa afirma que 80% de sus clientes están satisfechos. En una encuesta de 200 clientes, 150 reportan satisfacción. ¿Hay evidencia de que la proporción es menor? (α = 0.05)

**Paso 1: Hipótesis**
* H₀: p ≥ 0.80
* H₁: p < 0.80 (unilateral izquierda)

**Paso 2: Nivel de significancia**
* α = 0.05

**Paso 3: Estadístico**
* p̂ = 150/200 = 0.75

$$Z = \frac{0.75 - 0.80}{\sqrt{\frac{0.80 \times 0.20}{200}}} = \frac{-0.05}{0.0283} = -1.77$$

**Paso 4: Valor crítico**
* z_{0.05} = -1.645 (unilateral izquierda)

**Paso 5: Decisión**
* -1.77 < -1.645 → **Rechazar H₀**

**Paso 6: Conclusión**
* Hay evidencia suficiente para concluir que menos del 80% de los clientes están satisfechos.

## 6. Prueba t para Diferencia de Medias (Dos Muestras)

### Muestras Independientes

**Hipótesis:**
```
H₀: μ₁ = μ₂  (o μ₁ - μ₂ = 0)
H₁: μ₁ ≠ μ₂  (o μ₁ > μ₂, o μ₁ < μ₂)
```

**Estadístico (cuando n₁, n₂ ≥ 30):**
$$t = \frac{(\bar{x}_1 - \bar{x}_2)}{\sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}}$$

Grados de libertad aproximados (fórmula de Welch).

### Ejemplo

Comparar ventas promedio de dos sucursales:

**Sucursal A**: x̄₁ = $5,200, s₁ = $800, n₁ = 50
**Sucursal B**: x̄₂ = $4,800, s₂ = $750, n₂ = 45

¿Hay diferencia significativa? (α = 0.05)

**Paso 1: Hipótesis**
* H₀: μ₁ = μ₂
* H₁: μ₁ ≠ μ₂

**Paso 3: Estadístico**
$$t = \frac{5200 - 4800}{\sqrt{\frac{800^2}{50} + \frac{750^2}{45}}} = \frac{400}{157.4} = 2.54$$

**Conclusión**: Si t_{crítico} ≈ 2.00 con gl apropiados, entonces 2.54 > 2.00 → Rechazar H₀. Las ventas promedio son significativamente diferentes.

## 7. Regresión Lineal Simple

### Modelo

La **regresión lineal** modela la relación entre una variable dependiente Y y una variable independiente X:

$$Y = \beta_0 + \beta_1 X + \varepsilon$$

Donde:
* Y: Variable dependiente (respuesta)
* X: Variable independiente (predictora)
* β₀: Intercepto (ordenada al origen)
* β₁: Pendiente (cambio en Y por unidad de X)
* ε: Error aleatorio

### Estimación por Mínimos Cuadrados

El método de **mínimos cuadrados** minimiza la suma de los errores al cuadrado:

$$\text{Minimizar: } SSE = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

**Fórmulas de los estimadores:**

$$\hat{\beta}_1 = \frac{\sum_{i=1}^{n}(x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^{n}(x_i - \bar{x})^2}$$

$$\hat{\beta}_0 = \bar{y} - \hat{\beta}_1\bar{x}$$

### Ecuación de Regresión Estimada

$$\hat{y} = \hat{\beta}_0 + \hat{\beta}_1 x$$

Donde ŷ es el valor predicho de Y.

### Interpretación

* **β̂₀**: Valor esperado de Y cuando X = 0
* **β̂₁**: Cambio esperado en Y por cada unidad de aumento en X

### Ejemplo

Relación entre gasto en publicidad (X, miles $) y ventas (Y, miles $):

| Publicidad (X) | Ventas (Y) |
|---------------|------------|
| 1 | 10 |
| 2 | 15 |
| 3 | 18 |
| 4 | 22 |
| 5 | 25 |

Resultados:
* x̄ = 3, ȳ = 18
* β̂₁ = 3.7
* β̂₀ = 6.9

**Ecuación**: ŷ = 6.9 + 3.7x

**Interpretación**: Por cada $1,000 adicionales en publicidad, las ventas aumentan en promedio $3,700.

## 8. Coeficiente de Determinación (R²)

### Definición

El **coeficiente de determinación** R² mide la proporción de la variabilidad de Y que es explicada por el modelo de regresión.

### Fórmula

$$R^2 = \frac{SSR}{SST} = 1 - \frac{SSE}{SST}$$

Donde:
* **SST** (Total): $\sum(y_i - \bar{y})^2$ = Variabilidad total de Y
* **SSR** (Regresión): $\sum(\hat{y}_i - \bar{y})^2$ = Variabilidad explicada por X
* **SSE** (Error): $\sum(y_i - \hat{y}_i)^2$ = Variabilidad no explicada

**Relación**: SST = SSR + SSE

### Interpretación

* R² está entre 0 y 1
* **R² = 0.75** significa que el 75% de la variabilidad de Y es explicada por X
* **R² cercano a 1**: Buen ajuste del modelo
* **R² cercano a 0**: X no explica bien a Y

⚠️ **Cuidado**: R² alto no implica causalidad, puede haber correlaciones espurias.

### Ejemplo

En el ejemplo de publicidad y ventas:
* SST = 140
* SSR = 127.3
* SSE = 12.7

$$R^2 = \frac{127.3}{140} = 0.909$$

**Interpretación**: El 90.9% de la variabilidad en ventas es explicada por el gasto en publicidad.

## 9. Correlación

### Coeficiente de Correlación de Pearson (r)

Mide la **fuerza y dirección** de la relación lineal entre dos variables.

**Fórmula:**
$$r = \frac{\sum_{i=1}^{n}(x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum_{i=1}^{n}(x_i - \bar{x})^2 \sum_{i=1}^{n}(y_i - \bar{y})^2}}$$

O equivalente:
$$r = \frac{Cov(X,Y)}{\sigma_X \sigma_Y}$$

### Propiedades

* **Rango**: -1 ≤ r ≤ 1
* **r > 0**: Correlación positiva (X aumenta → Y aumenta)
* **r < 0**: Correlación negativa (X aumenta → Y disminuye)
* **r = 0**: No correlación lineal
* **|r| cercano a 1**: Correlación fuerte
* **|r| cercano a 0**: Correlación débil

### Interpretación de |r|

| |r| | Interpretación |
|-----|----------------|
| 0.00 - 0.30 | Débil |
| 0.30 - 0.70 | Moderada |
| 0.70 - 1.00 | Fuerte |

### Relación con R²

En regresión lineal simple:
$$R^2 = r^2$$

### Prueba de Significancia

Para probar si hay correlación en la población:

**Hipótesis:**
```
H₀: ρ = 0 (no hay correlación)
H₁: ρ ≠ 0
```

**Estadístico:**
$$t = \frac{r\sqrt{n-2}}{\sqrt{1-r^2}}$$

con n-2 grados de libertad.

### Cuidados

⚠️ **Correlación NO implica causalidad**

* Puede haber variables confusoras
* La relación puede ser espuria
* Solo mide relación lineal (puede haber relaciones no lineales)

## 10. Análisis de Varianza (ANOVA)

### Objetivo

Comparar **medias de tres o más grupos** independientes simultáneamente.

**¿Por qué no múltiples pruebas t?**
* Aumenta el error Tipo I
* ANOVA controla el error global

### Hipótesis

```
H₀: μ₁ = μ₂ = μ₃ = ... = μₖ
    (todas las medias son iguales)
    
H₁: Al menos una media es diferente
```

### Lógica del ANOVA

Compara:
* **Variabilidad entre grupos** (debido a tratamientos)
* **Variabilidad dentro de grupos** (debido a error aleatorio)

Si la variabilidad entre grupos es mucho mayor que dentro de grupos, rechazamos H₀.

### Tabla ANOVA

| Fuente | Suma de Cuadrados | gl | Cuadrado Medio | F |
|--------|------------------|-----|----------------|-----|
| Entre grupos | SSB | k-1 | MSB = SSB/(k-1) | F = MSB/MSW |
| Dentro grupos | SSW | N-k | MSW = SSW/(N-k) | |
| Total | SST | N-1 | | |

Donde:
* k = número de grupos
* N = tamaño total de muestra
* gl = grados de libertad

### Estadístico F

$$F = \frac{MSB}{MSW} = \frac{\text{Variabilidad entre grupos}}{\text{Variabilidad dentro de grupos}}$$

**Distribución**: F con gl₁ = k-1 y gl₂ = N-k

**Decisión:**
* Si F > F_{crítico} (o p-valor < α): Rechazar H₀
* Conclusión: Al menos dos medias son diferentes

### Supuestos

1. **Independencia**: Observaciones independientes
2. **Normalidad**: Cada grupo sigue distribución normal
3. **Homocedasticidad**: Varianzas iguales entre grupos

### Ejemplo

Comparar ventas promedio de 3 estrategias de marketing:

**Estrategia A**: [120, 125, 118, 130, 122]
**Estrategia B**: [135, 140, 138, 142, 137]
**Estrategia C**: [115, 118, 120, 116, 121]

Resultados del ANOVA:

| Fuente | SS | gl | MS | F | p-valor |
|--------|----|----|-------|-----|--------|
| Entre | 1500 | 2 | 750 | 45.5 | <0.001 |
| Dentro | 198 | 12 | 16.5 | | |
| Total | 1698 | 14 | | | |

**Conclusión**: F = 45.5 con p < 0.001 → Rechazar H₀. Hay diferencia significativa entre las estrategias.

### Post-hoc

Si rechazamos H₀, necesitamos **pruebas post-hoc** (ej: Tukey, Bonferroni) para identificar qué pares de grupos difieren.

## 11. Herramientas en Python

### Pruebas de Hipótesis

```python
from scipy import stats
import numpy as np

# Prueba t de una muestra
data = [3.2, 3.5, 3.1, 3.8, 3.4, 3.6]
t_stat, p_value = stats.ttest_1samp(data, popmean=3.0)
print(f"t = {t_stat:.3f}, p-valor = {p_value:.4f}")

# Prueba t de dos muestras independientes
group1 = [5200, 5400, 5100, ...]
group2 = [4800, 5000, 4700, ...]
t_stat, p_value = stats.ttest_ind(group1, group2)

# Prueba Z para proporción
from statsmodels.stats.proportion import proportions_ztest
count = 150  # éxitos
nobs = 200   # tamaño muestra
z_stat, p_value = proportions_ztest(count, nobs, value=0.80)
```

### Regresión Lineal

```python
from scipy import stats
import matplotlib.pyplot as plt

# Datos
x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 15, 18, 22, 25])

# Regresión
slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)

print(f"Ecuación: y = {intercept:.2f} + {slope:.2f}x")
print(f"R² = {r_value**2:.4f}")
print(f"p-valor = {p_value:.4f}")

# Predicciones
y_pred = slope * x + intercept

# Visualización
plt.scatter(x, y, label='Datos')
plt.plot(x, y_pred, 'r-', label='Regresión')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```

### Correlación

```python
# Correlación de Pearson
r, p_value = stats.pearsonr(x, y)
print(f"r = {r:.4f}, p-valor = {p_value:.4f}")

# Correlación de Spearman (no paramétrico)
rho, p_value = stats.spearmanr(x, y)
```

### ANOVA

```python
from scipy import stats

# Datos de 3 grupos
group1 = [120, 125, 118, 130, 122]
group2 = [135, 140, 138, 142, 137]
group3 = [115, 118, 120, 116, 121]

# ANOVA de un factor
F_stat, p_value = stats.f_oneway(group1, group2, group3)

print(f"F = {F_stat:.2f}, p-valor = {p_value:.4f}")

if p_value < 0.05:
    print("Rechazar H₀: Hay diferencias significativas")
else:
    print("No rechazar H₀: No hay diferencias significativas")
```

### Regresión con statsmodels (más completo)

```python
import statsmodels.api as sm

# Preparar datos
X = sm.add_constant(x)  # Añadir intercepto

# Ajustar modelo
model = sm.OLS(y, X)
results = model.fit()

# Resumen completo
print(results.summary())

# Coeficientes e intervalos de confianza
print(results.params)
print(results.conf_int())
```

## Resumen de Fórmulas Clave

### Pruebas de Hipótesis
* Estadístico Z: (x̄ - μ₀)/(σ/√n)
* Estadístico t: (x̄ - μ₀)/(s/√n)
* Proporción Z: (p̂ - p₀)/√[p₀(1-p₀)/n]

### Regresión
* Pendiente: β̂₁ = Σ[(xᵢ-x̄)(yᵢ-ȳ)]/Σ(xᵢ-x̄)²
* Intercepto: β̂₀ = ȳ - β̂₁x̄
* R²: 1 - SSE/SST

### Correlación
* r = Cov(X,Y)/(σₓσᵧ)
* Prueba t: t = r√(n-2)/√(1-r²)

### ANOVA
* F = MSB/MSW
* MSB = SSB/(k-1)
* MSW = SSW/(N-k)
