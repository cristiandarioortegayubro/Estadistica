## Ejercicio 1: Evaluación de Satisfacción del Cliente

### Problema de Negocios
Una empresa evalúa la satisfacción del cliente en dos dimensiones independientes (escala 1-6):
* **Calidad del Producto (P)**: calificación de 1 a 6
* **Calidad del Servicio (S)**: calificación de 1 a 6

Cada calificación es equiprobable (como lanzar un dado justo).

**Preguntas:**
1. ¿Cuál es la probabilidad de que la satisfacción total sea exactamente 7 (cliente satisfecho)?
2. ¿Cuál es la probabilidad de que la satisfacción total sea ≥ 10 (cliente muy satisfecho)?
3. ¿Cuál es la probabilidad de que al menos una dimensión tenga calificación máxima (6)?

Vamos a resolver esto teóricamente y verificar con simulación.

### 📚 Conceptos Fundamentales

**Probabilidad Clásica (Laplace)**

Cuando todos los resultados son igualmente probables:

$$P(A) = \frac{\text{Número de casos favorables}}{\text{Número total de casos posibles}}$$

**Espacio Muestral (Ω)**
* Es el conjunto de TODOS los resultados posibles de un experimento
* En nuestro problema: Ω = {(1,1), (1,2), ..., (6,6)} → 36 combinaciones
* Cada combinación (P, S) representa una calificación posible

**Ejemplo:**
* Si P=3 y S=4, el cliente dio 3 en Producto y 4 en Servicio
* La satisfacción total sería 3+4 = 7

**Evento (A)**
* Es un subconjunto del espacio muestral
* Ejemplo: A = "Satisfacción total = 7" = {(1,6), (2,5), (3,4), (4,3), (5,2), (6,1)}
* Hay 6 formas de obtener satisfacción total = 7

**Regla del Producto**
* Si tenemos dos dimensiones independientes con 6 valores cada una:
* Total de combinaciones = 6 × 6 = 36

```python
# ============================================================================
# PASO 1: Construir el espacio muestral (Ω)
# ============================================================================
# El espacio muestral es el conjunto de TODOS los resultados posibles.
# En este caso: todas las combinaciones de (Producto, Servicio) donde cada uno va de 1 a 6
espacio_muestral = [(p, s) for p in range(1, 7) for s in range(1, 7)]
total_resultados = len(espacio_muestral)  # Debe ser 6 × 6 = 36 combinaciones

print("=== Análisis de Satisfacción del Cliente ===")
print(f"\nTotal de combinaciones posibles: {total_resultados}")
print(f"Primeras 10 combinaciones (Producto, Servicio): {espacio_muestral[:10]}")

# ============================================================================
# PREGUNTA 1: Satisfacción total = 7 (cliente satisfecho)
# ============================================================================
# Identificamos el evento A = "satisfacción total = 7"
# Filtramos todas las combinaciones donde p + s = 7
satisfaccion_7 = [(p, s) for p, s in espacio_muestral if p + s == 7]

# Aplicamos probabilidad clásica: P(A) = casos favorables / casos totales
prob_satisfaccion_7 = len(satisfaccion_7) / total_resultados

print(f"\n1. Satisfacción Total = 7 (Cliente Satisfecho):")
print(f"   Combinaciones posibles: {satisfaccion_7}")
print(f"   P(Satisfacción = 7) = {len(satisfaccion_7)}/{total_resultados} = {prob_satisfaccion_7:.4f} ({prob_satisfaccion_7*100:.1f}%)")

# ============================================================================
# PREGUNTA 2: Satisfacción total ≥ 10 (cliente muy satisfecho)
# ============================================================================
# Identificamos el evento B = "satisfacción total ≥ 10"
# Filtramos todas las combinaciones donde p + s ≥ 10
satisfaccion_alta = [(p, s) for p, s in espacio_muestral if p + s >= 10]

# Calculamos la probabilidad: P(B) = |B| / |Ω|
prob_satisfaccion_alta = len(satisfaccion_alta) / total_resultados

print(f"\n2. Satisfacción Total ≥ 10 (Cliente Muy Satisfecho):")
print(f"   Combinaciones posibles: {satisfaccion_alta}")
print(f"   P(Satisfacción ≥ 10) = {len(satisfaccion_alta)}/{total_resultados} = {prob_satisfaccion_alta:.4f} ({prob_satisfaccion_alta*100:.1f}%)")

# ============================================================================
# PREGUNTA 3: Al menos una dimensión con calificación máxima (6)
# ============================================================================
# Identificamos el evento C = "P=6 O S=6" (unión de eventos)
# Usamos el operador OR (or) para incluir casos donde:
#   - Producto = 6 (independientemente del servicio)
#   - Servicio = 6 (independientemente del producto)
#   - Ambos = 6 (incluido automáticamente por el OR)
calificacion_maxima = [(p, s) for p, s in espacio_muestral if p == 6 or s == 6]

# Calculamos la probabilidad: P(C) = |C| / |Ω|
prob_calificacion_maxima = len(calificacion_maxima) / total_resultados

print(f"\n3. Al Menos Una Dimensión con Calificación Máxima (6):")
print(f"   Casos favorables: {len(calificacion_maxima)}")
print(f"   P(Al menos un 6) = {len(calificacion_maxima)}/{total_resultados} = {prob_calificacion_maxima:.4f} ({prob_calificacion_maxima*100:.1f}%)")

print(f"\nℹ️ Interpretación de Negocios:")
print(f"   - {prob_satisfaccion_7*100:.1f}% de los clientes tienen satisfacción moderada (total = 7)")
print(f"   - {prob_satisfaccion_alta*100:.1f}% de los clientes están muy satisfechos (total ≥ 10)")
print(f"   - {prob_calificacion_maxima*100:.1f}% dan calificación perfecta en al menos una dimensión")
```

```python
# Simulación: evaluar satisfacción de muchos clientes
n_simulaciones = 100000

print(f"\n=== Simulación con {n_simulaciones:,} clientes ===")

# Simular calificaciones de clientes
calif_producto = np.random.randint(1, 7, n_simulaciones)
calif_servicio = np.random.randint(1, 7, n_simulaciones)
satisfaccion_total = calif_producto + calif_servicio

# Calcular frecuencias relativas
freq_satisfaccion_7 = np.sum(satisfaccion_total == 7) / n_simulaciones
freq_satisfaccion_alta = np.sum(satisfaccion_total >= 10) / n_simulaciones
freq_calificacion_maxima = np.sum((calif_producto == 6) | (calif_servicio == 6)) / n_simulaciones

print(f"\n📊 Comparación: Probabilidades Teóricas vs Simuladas")
print(f"\n1. P(Satisfacción Total = 7):")
print(f"   Teórica:    {prob_satisfaccion_7:.4f} ({prob_satisfaccion_7*100:.1f}%)")
print(f"   Simulada:   {freq_satisfaccion_7:.4f} ({freq_satisfaccion_7*100:.1f}%)")
print(f"   Diferencia: {abs(prob_satisfaccion_7 - freq_satisfaccion_7):.4f}")

print(f"\n2. P(Satisfacción Total ≥ 10):")
print(f"   Teórica:    {prob_satisfaccion_alta:.4f} ({prob_satisfaccion_alta*100:.1f}%)")
print(f"   Simulada:   {freq_satisfaccion_alta:.4f} ({freq_satisfaccion_alta*100:.1f}%)")
print(f"   Diferencia: {abs(prob_satisfaccion_alta - freq_satisfaccion_alta):.4f}")

print(f"\n3. P(Al menos una calificación de 6):")
print(f"   Teórica:    {prob_calificacion_maxima:.4f} ({prob_calificacion_maxima*100:.1f}%)")
print(f"   Simulada:   {freq_calificacion_maxima:.4f} ({freq_calificacion_maxima*100:.1f}%)")
print(f"   Diferencia: {abs(prob_calificacion_maxima - freq_calificacion_maxima):.4f}")

# Visualización de la distribución de satisfacción
plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.hist(satisfaccion_total, bins=np.arange(2, 14) - 0.5, edgecolor='black', density=True, color='#3498db', alpha=0.7)
plt.xlabel('Satisfacción Total (Producto + Servicio)', fontsize=11)
plt.ylabel('Frecuencia Relativa', fontsize=11)
plt.title('Distribución de Satisfacción Total - Simulación', fontsize=12, fontweight='bold')
plt.xticks(range(2, 13))
plt.grid(axis='y', alpha=0.3)

# Marcar zonas de satisfacción
plt.axvline(x=7, color='orange', linestyle='--', linewidth=2, label='Satisfecho (=7)')
plt.axvline(x=10, color='green', linestyle='--', linewidth=2, label='Muy Satisfecho (≥10)')
plt.legend()

plt.subplot(1, 2, 2)
# Distribución teórica
satisfacciones_teoricas = [p + s for p, s in espacio_muestral]
contador = Counter(satisfacciones_teoricas)
satisfacciones_unicas = sorted(contador.keys())
probs_teoricas = [contador[s] / total_resultados for s in satisfacciones_unicas]

colores = ['#e74c3c' if s < 7 else '#f39c12' if s == 7 else '#2ecc71' if s >= 10 else '#95a5a6' 
           for s in satisfacciones_unicas]
plt.bar(satisfacciones_unicas, probs_teoricas, edgecolor='black', alpha=0.7, color=colores)
plt.xlabel('Satisfacción Total (Producto + Servicio)', fontsize=11)
plt.ylabel('Probabilidad', fontsize=11)
plt.title('Distribución Teórica de Satisfacción', fontsize=12, fontweight='bold')
plt.xticks(range(2, 13))
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\n✅ Conclusión: La simulación valida los cálculos teóricos.")
print(f"   Las probabilidades simuladas coinciden con las teóricas dentro del margen de error esperado.")
```

## Ejercicio 2: Análisis de Clientes (Regla de la Suma)

### Problema
Una empresa analiza su base de clientes y encuentra que:
* 40% tienen membresía Premium (P)
* 30% han realizado una compra en el último mes (C)
* 15% tienen membresía Premium Y compraron en el último mes

Calcular:
1. Probabilidad de que un cliente tenga membresía Premium O haya comprado en el último mes
2. Probabilidad de que no tenga membresía Premium
3. Probabilidad de que tenga membresía Premium pero NO haya comprado

### 📚 Conceptos Fundamentales

**Regla de la Suma (Unión de Eventos)**

Para calcular la probabilidad de que ocurra A **O** B:

$$P(A \cup B) = P(A) + P(B) - P(A \cap B)$$

**¿Por qué restamos la intersección?**
* Si solo sumamos P(A) + P(B), estaríamos contando dos veces los casos donde ocurren ambos
* La intersección P(A ∩ B) representa los casos que cumplen ambas condiciones simultáneamente

**Casos especiales:**
* **Eventos mutuamente excluyentes**: Si A y B no pueden ocurrir al mismo tiempo, entonces P(A ∩ B) = 0
  * En ese caso: P(A ∪ B) = P(A) + P(B)

**Regla del Complemento**

La probabilidad de que NO ocurra A:

$$P(A^c) = 1 - P(A)$$

**Ejemplo en nuestro problema:**
* A = "Cliente Premium" (40%)
* B = "Compró último mes" (30%)
* A ∩ B = "Premium Y compró" (15%)
* A ∪ B = 40% + 30% - 15% = 55%

```python
# ============================================================================
# DATOS DEL PROBLEMA
# ============================================================================
# Definimos las probabilidades conocidas:
P_premium = 0.40              # P(P) = Probabilidad de tener membresía Premium
P_compra = 0.30               # P(C) = Probabilidad de haber comprado último mes
P_premium_y_compra = 0.15     # P(P ∩ C) = Probabilidad de AMBAS (intersección)

print("=== Análisis de Clientes ===")

# ============================================================================
# PREGUNTA 1: P(Premium O Compra) - Aplicamos la REGLA DE LA SUMA
# ============================================================================
# Fórmula: P(A ∪ B) = P(A) + P(B) - P(A ∩ B)
# Restamos la intersección para no contar dos veces los que cumplen ambas condiciones
P_premium_o_compra = P_premium + P_compra - P_premium_y_compra

print(f"\n1. P(Premium ∪ Compra) = P(P) + P(C) - P(P ∩ C)")
print(f"   = {P_premium} + {P_compra} - {P_premium_y_compra}")
print(f"   = {P_premium_o_compra:.2f} ({P_premium_o_compra*100:.0f}%)")
print(f"   \u2192 El {P_premium_o_compra*100:.0f}% de los clientes tiene Premium O compró (o ambos)")

# ============================================================================
# PREGUNTA 2: P(No Premium) - Aplicamos la REGLA DEL COMPLEMENTO
# ============================================================================
# Fórmula: P(Aᶜ) = 1 - P(A)
# El complemento de "tener Premium" es "NO tener Premium"
P_no_premium = 1 - P_premium

print(f"\n2. P(No Premium) = 1 - P(Premium)")
print(f"   = 1 - {P_premium}")
print(f"   = {P_no_premium:.2f} ({P_no_premium*100:.0f}%)")
print(f"   \u2192 El {P_no_premium*100:.0f}% de los clientes NO tiene membresía Premium")

# ============================================================================
# PREGUNTA 3: P(Premium y NO Compra)
# ============================================================================
# Los que tienen Premium se dividen en dos grupos:
#   - Los que tienen Premium Y compraron: P(P ∩ C) = 0.15
#   - Los que tienen Premium Y NO compraron: P(P ∩ Cᶜ) = ?
# Por lo tanto: P(P ∩ Cᶜ) = P(P) - P(P ∩ C)
P_premium_y_no_compra = P_premium - P_premium_y_compra

print(f"\n3. P(Premium ∩ No Compra) = P(Premium) - P(Premium ∩ Compra)")
print(f"   = {P_premium} - {P_premium_y_compra}")
print(f"   = {P_premium_y_no_compra:.2f} ({P_premium_y_no_compra*100:.0f}%)")
print(f"   \u2192 El {P_premium_y_no_compra*100:.0f}% tiene Premium pero NO compró")

# Visualización con diagrama de Venn (simulado)
fig, ax = plt.subplots(1, 1, figsize=(10, 6))

# Crear datos simulados para visualización
n_clientes = 1000
premium = int(P_premium * n_clientes)
compra = int(P_compra * n_clientes)
premium_y_compra = int(P_premium_y_compra * n_clientes)

categorias = ['Solo Premium', 'Premium y Compra', 'Solo Compra', 'Ninguno']
valores = [
    premium - premium_y_compra,  # Solo Premium
    premium_y_compra,             # Ambos
    compra - premium_y_compra,    # Solo Compra
    n_clientes - (premium + compra - premium_y_compra)  # Ninguno
]

colores = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99']
ax.pie(valores, labels=categorias, autopct='%1.1f%%', startangle=90, colors=colores)
ax.set_title('Distribución de Clientes', fontsize=14, fontweight='bold')
plt.axis('equal')
plt.show()

print(f"\nℹ️ Interpretación:")
print(f"   - {valores[0]/n_clientes*100:.1f}% tienen membresía Premium pero no compraron")
print(f"   - {valores[1]/n_clientes*100:.1f}% tienen Premium Y compraron (intersección)")
print(f"   - {valores[2]/n_clientes*100:.1f}% compraron pero no tienen Premium")
print(f"   - {valores[3]/n_clientes*100:.1f}% no tienen Premium ni compraron")
```

## Ejercicio 3: Control de Calidad (Probabilidad Condicional)

### Problema
Una fábrica produce componentes en dos turnos:
* Turno mañana (M): 60% de la producción
* Turno tarde (T): 40% de la producción

Tasas de defectos:
* Mañana: 2% de defectos
* Tarde: 5% de defectos

Calcular:
1. ¿Cuál es la probabilidad de que un componente sea defectuoso?
2. Si un componente es defectuoso, ¿cuál es la probabilidad de que provenga del turno tarde?
3. Si un componente es bueno, ¿cuál es la probabilidad de que sea del turno mañana?

```python
# Datos del problema
P_manana = 0.60
P_tarde = 0.40
P_defecto_dado_manana = 0.02
P_defecto_dado_tarde = 0.05

print("=== Control de Calidad ===")
print(f"\nDatos:")
print(f"  P(Mañana) = {P_manana}")
print(f"  P(Tarde) = {P_tarde}")
print(f"  P(Defecto | Mañana) = {P_defecto_dado_manana}")
print(f"  P(Defecto | Tarde) = {P_defecto_dado_tarde}")

# 1. P(Defecto) - Probabilidad total
P_defecto = P_manana * P_defecto_dado_manana + P_tarde * P_defecto_dado_tarde

print(f"\n1. P(Defecto) = P(M) × P(D|M) + P(T) × P(D|T)")
print(f"   = {P_manana} × {P_defecto_dado_manana} + {P_tarde} × {P_defecto_dado_tarde}")
print(f"   = {P_defecto:.4f} ({P_defecto*100:.2f}%)")

# 2. P(Tarde | Defecto) - Teorema de Bayes
P_tarde_dado_defecto = (P_tarde * P_defecto_dado_tarde) / P_defecto

print(f"\n2. P(Tarde | Defecto) - Teorema de Bayes")
print(f"   = P(T) × P(D|T) / P(D)")
print(f"   = ({P_tarde} × {P_defecto_dado_tarde}) / {P_defecto:.4f}")
print(f"   = {P_tarde_dado_defecto:.4f} ({P_tarde_dado_defecto*100:.2f}%)")

# 3. P(Mañana | Bueno)
P_bueno_dado_manana = 1 - P_defecto_dado_manana
P_bueno_dado_tarde = 1 - P_defecto_dado_tarde
P_bueno = 1 - P_defecto

P_manana_dado_bueno = (P_manana * P_bueno_dado_manana) / P_bueno

print(f"\n3. P(Mañana | Bueno)")
print(f"   P(Bueno) = {P_bueno:.4f}")
print(f"   P(Bueno | Mañana) = {P_bueno_dado_manana}")
print(f"   P(Mañana | Bueno) = {P_manana_dado_bueno:.4f} ({P_manana_dado_bueno*100:.2f}%)")

print(f"\nℹ️ Conclusión: Aunque el turno tarde produce menos (40%), el {P_tarde_dado_defecto*100:.1f}% de")
print(f"   los componentes defectuosos provienen de ese turno debido a su mayor tasa de defectos.")
```

```python
# Simular producción de componentes
n_componentes = 10000

# Asignar turnos según probabilidades
turnos = np.random.choice(['Mañana', 'Tarde'], size=n_componentes, p=[P_manana, P_tarde])

# Generar defectos según el turno
defectos = []
for turno in turnos:
    if turno == 'Mañana':
        es_defecto = np.random.random() < P_defecto_dado_manana
    else:
        es_defecto = np.random.random() < P_defecto_dado_tarde
    defectos.append(es_defecto)

defectos = np.array(defectos)

# Crear DataFrame para análisis
df = pd.DataFrame({
    'Turno': turnos,
    'Defecto': defectos
})

print("\n=== Resultados de la simulación ===")
print(f"\nTotal de componentes simulados: {n_componentes:,}")

# Tabla de frecuencias
print("\nTabla de frecuencias:")
tabla = pd.crosstab(df['Turno'], df['Defecto'], margins=True)
tabla.columns = ['Bueno', 'Defecto', 'Total']
print(tabla)

# Probabilidades simuladas
print("\n\nProbabilidades simuladas vs teóricas:")
print(f"\n1. P(Defecto):")
prob_defecto_sim = defectos.mean()
print(f"   Teórica:  {P_defecto:.4f}")
print(f"   Simulada: {prob_defecto_sim:.4f}")
print(f"   Error:    {abs(P_defecto - prob_defecto_sim):.4f}")

# P(Tarde | Defecto)
defectos_tarde = np.sum((df['Turno'] == 'Tarde') & df['Defecto'])
total_defectos = np.sum(df['Defecto'])
prob_tarde_dado_defecto_sim = defectos_tarde / total_defectos if total_defectos > 0 else 0

print(f"\n2. P(Tarde | Defecto):")
print(f"   Teórica:  {P_tarde_dado_defecto:.4f}")
print(f"   Simulada: {prob_tarde_dado_defecto_sim:.4f}")
print(f"   Error:    {abs(P_tarde_dado_defecto - prob_tarde_dado_defecto_sim):.4f}")

# Visualización
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Gráfico 1: Distribución por turno
turno_counts = df['Turno'].value_counts()
axes[0].bar(turno_counts.index, turno_counts.values, color=['#3498db', '#e74c3c'], edgecolor='black')
axes[0].set_xlabel('Turno')
axes[0].set_ylabel('Número de componentes')
axes[0].set_title('Distribución de Producción por Turno')
axes[0].grid(axis='y', alpha=0.3)

# Agregar valores sobre las barras
for i, v in enumerate(turno_counts.values):
    axes[0].text(i, v + 100, f'{v:,}\n({v/n_componentes*100:.1f}%)', ha='center')

# Gráfico 2: Tasa de defectos por turno
tasas_defecto = df.groupby('Turno')['Defecto'].mean() * 100
colores_turno = ['#3498db', '#e74c3c']
axes[1].bar(tasas_defecto.index, tasas_defecto.values, color=colores_turno, edgecolor='black')
axes[1].set_xlabel('Turno')
axes[1].set_ylabel('% Defectos')
axes[1].set_title('Tasa de Defectos por Turno')
axes[1].grid(axis='y', alpha=0.3)

# Agregar valores sobre las barras
for i, (turno, tasa) in enumerate(tasas_defecto.items()):
    axes[1].text(i, tasa + 0.1, f'{tasa:.2f}%', ha='center')

plt.tight_layout()
plt.show()
```

## Ejercicio 4: Análisis de Independencia

### Problema
Una empresa de e-commerce analiza dos características de sus clientes:
* A: Cliente usa la app móvil
* B: Cliente realiza compras frecuentes (más de 5 al mes)

Datos observados de 1000 clientes:
* 400 usan la app móvil
* 300 realizan compras frecuentes
* 150 usan la app Y realizan compras frecuentes

**Pregunta:** ¿Son independientes estos sucesos?

### 📚 Conceptos Fundamentales

**Independencia Estadística**

Dos eventos A y B son **independientes** si el hecho de que ocurra uno NO afecta la probabilidad del otro.

**Definición matemática** (cualquiera de estas tres condiciones equivalentes):

1. $$P(A \cap B) = P(A) \cdot P(B)$$
2. $$P(B|A) = P(B)$$
3. $$P(A|B) = P(A)$$

**Interpretación:**
* Si son independientes: saber que A ocurrió no cambia la probabilidad de B
* Si NO son independientes: hay una relación (positiva o negativa) entre A y B

**Ejemplo de independencia:**
* Lanzar dos dados: el resultado del primer dado NO afecta el resultado del segundo
* P(Dado1 = 6 ∩ Dado2 = 6) = P(Dado1 = 6) × P(Dado2 = 6) = 1/6 × 1/6 = 1/36 ✓

**Ejemplo de dependencia:**
* Sacar dos cartas de un mazo SIN reposición
* P(Segunda carta es As | Primera carta fue As) ≠ P(Segunda carta es As)
* Porque ya hay un As menos en el mazo

**En nuestro problema:**
* Verificaremos si "Usar la app" y "Comprar frecuentemente" son independientes
* Si P(Compra Frecuente | Usa App) ≠ P(Compra Frecuente), entonces NO son independientes

```python
# Datos observados
n_total = 1000
n_usa_app = 400
n_compra_frecuente = 300
n_ambos = 150

# Calcular probabilidades
P_A = n_usa_app / n_total
P_B = n_compra_frecuente / n_total
P_A_y_B = n_ambos / n_total

print("=== Análisis de Independencia ===")
print(f"\nProbabilidades observadas:")
print(f"  P(Usa App) = {P_A}")
print(f"  P(Compra Frecuente) = {P_B}")
print(f"  P(Usa App ∩ Compra Frecuente) = {P_A_y_B}")

# Si fueran independientes, debería cumplirse: P(A ∩ B) = P(A) × P(B)
P_A_y_B_si_indep = P_A * P_B

print(f"\n¿Son independientes?")
print(f"  Si fueran independientes: P(A ∩ B) = P(A) × P(B) = {P_A} × {P_B} = {P_A_y_B_si_indep}")
print(f"  Observado: P(A ∩ B) = {P_A_y_B}")
print(f"  Diferencia: {abs(P_A_y_B - P_A_y_B_si_indep):.4f}")

if abs(P_A_y_B - P_A_y_B_si_indep) < 0.01:
    print(f"\n✅ Los sucesos SON aproximadamente independientes")
else:
    print(f"\n❌ Los sucesos NO son independientes")

# Calcular probabilidades condicionales
P_B_dado_A = P_A_y_B / P_A
P_B_dado_no_A = (P_B - P_A_y_B) / (1 - P_A)

print(f"\nProbabilidades condicionales:")
print(f"  P(Compra Frecuente | Usa App) = {P_B_dado_A:.4f}")
print(f"  P(Compra Frecuente | No Usa App) = {P_B_dado_no_A:.4f}")
print(f"  P(Compra Frecuente) = {P_B}")

if abs(P_B_dado_A - P_B) < 0.01:
    print(f"\n✅ P(B|A) ≈ P(B) → Confirma independencia")
else:
    diferencia_rel = ((P_B_dado_A - P_B) / P_B) * 100
    if P_B_dado_A > P_B:
        print(f"\n📈 Usar la app AUMENTA la probabilidad de compras frecuentes en {diferencia_rel:.1f}%")
    else:
        print(f"\n📉 Usar la app DISMINUYE la probabilidad de compras frecuentes en {abs(diferencia_rel):.1f}%")

print(f"\nℹ️ Interpretación: {P_B_dado_A*100:.1f}% de quienes usan la app compran frecuentemente,")
print(f"   vs {P_B_dado_no_A*100:.1f}% de quienes NO usan la app. Hay una relación positiva.")
```

```python
# Crear tabla de contingencia
data_tabla = {
    'Usa App': [n_ambos, n_usa_app - n_ambos],
    'No Usa App': [n_compra_frecuente - n_ambos, n_total - n_usa_app - n_compra_frecuente + n_ambos]
}
index_tabla = ['Compra Frecuente', 'No Compra Frecuente']

df_contingencia = pd.DataFrame(data_tabla, index=index_tabla)

print("\n=== Tabla de Contingencia ===")
print(df_contingencia)
print(f"\nTotal: {df_contingencia.sum().sum()}")

# Visualización
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Gráfico 1: Tabla de contingencia como heatmap
im = axes[0].imshow(df_contingencia.values, cmap='Blues', aspect='auto')
axes[0].set_xticks(range(len(df_contingencia.columns)))
axes[0].set_yticks(range(len(df_contingencia.index)))
axes[0].set_xticklabels(df_contingencia.columns)
axes[0].set_yticklabels(df_contingencia.index)
axes[0].set_title('Tabla de Contingencia')

# Agregar valores en cada celda
for i in range(len(df_contingencia.index)):
    for j in range(len(df_contingencia.columns)):
        text = axes[0].text(j, i, df_contingencia.iloc[i, j],
                          ha="center", va="center", color="black", fontsize=14, fontweight='bold')

plt.colorbar(im, ax=axes[0])

# Gráfico 2: Comparación de probabilidades condicionales
categorias = ['P(B|A)', 'P(B|~A)', 'P(B)']
valores = [P_B_dado_A * 100, P_B_dado_no_A * 100, P_B * 100]
colores = ['#2ecc71' if P_B_dado_A > P_B else '#e74c3c', 
           '#e67e22' if P_B_dado_no_A < P_B else '#3498db', 
           '#95a5a6']

axes[1].bar(categorias, valores, color=colores, edgecolor='black')
axes[1].set_ylabel('Probabilidad (%)')
axes[1].set_title('Comparación de Probabilidades Condicionales')
axes[1].grid(axis='y', alpha=0.3)

# Agregar valores sobre las barras
for i, (cat, val) in enumerate(zip(categorias, valores)):
    axes[1].text(i, val + 1, f'{val:.1f}%', ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
```

## Ejercicio 5: Demostración de la Teoría Frecuencial

### Objetivo
Demostrar empíricamente que la frecuencia relativa converge a la probabilidad teórica cuando aumenta el número de repeticiones.

Vamos a lanzar una moneda muchas veces y observar cómo la frecuencia relativa de "Cara" se acerca a 0.5.

### 📚 Conceptos Fundamentales

**Ley de los Grandes Números (LGN)**

A medida que el número de repeticiones de un experimento aumenta, la **frecuencia relativa** de un evento converge a su **probabilidad teórica**.

$$\lim_{n \to \infty} \frac{\text{Número de éxitos}}{n} = P(\text{éxito})$$

**Frecuencia Relativa:**

$$f_n(A) = \frac{\text{Veces que ocurrió A en n repeticiones}}{n}$$

**Interpretación:**
* Con pocas repeticiones: la frecuencia relativa puede ser muy diferente de la probabilidad
* Con muchas repeticiones: la frecuencia relativa se estabiliza cerca de la probabilidad

**Ejemplo: Moneda justa**
* Probabilidad teórica: P(Cara) = 0.5
* Si lanzamos 10 veces: podríamos obtener 7 caras → f₁₀ = 0.7 (error = 0.2)
* Si lanzamos 1000 veces: podríamos obtener 503 caras → f₁₀₀₀ = 0.503 (error = 0.003)
* Si lanzamos 10000 veces: podríamos obtener 5012 caras → f₁₀₀₀₀ = 0.5012 (error = 0.0012)

**Implicación práctica:**
* Podemos ESTIMAR probabilidades desconocidas mediante experimentos repetidos
* Mientras más repeticiones, mejor será la estimación
* Esta es la base de la **simulación Monte Carlo**

```python
# Experimento: lanzar una moneda n veces
max_lanzamientos = 10000
lanzamientos = np.random.choice(['Cara', 'Cruz'], size=max_lanzamientos)

# Calcular frecuencia relativa acumulada
caras_acumuladas = np.cumsum(lanzamientos == 'Cara')
n_lanzamientos = np.arange(1, max_lanzamientos + 1)
frecuencia_relativa = caras_acumuladas / n_lanzamientos

# Probabilidad teórica
prob_teorica = 0.5

print("=== Demostración de la Teoría Frecuencial ===")
print(f"\nProbabilidad teórica de Cara: {prob_teorica}")
print(f"\nConvergencia de la frecuencia relativa:")

milestones = [10, 50, 100, 500, 1000, 5000, 10000]
for n in milestones:
    freq = frecuencia_relativa[n-1]
    error = abs(freq - prob_teorica)
    print(f"  Después de {n:5} lanzamientos: {freq:.4f} (error: {error:.4f})")

print(f"\nℹ️ Observación: A medida que aumenta n, la frecuencia relativa")
print(f"   se acerca cada vez más a la probabilidad teórica de 0.5")
```

```python
# Gráfico de convergencia
fig, axes = plt.subplots(2, 1, figsize=(14, 10))

# Gráfico 1: Frecuencia relativa vs número de lanzamientos
axes[0].plot(n_lanzamientos, frecuencia_relativa, label='Frecuencia relativa', linewidth=1.5)
axes[0].axhline(y=prob_teorica, color='r', linestyle='--', linewidth=2, label='Probabilidad teórica (0.5)')
axes[0].fill_between(n_lanzamientos, prob_teorica - 0.05, prob_teorica + 0.05, 
                     color='red', alpha=0.1, label='Margen ±5%')
axes[0].set_xlabel('Número de lanzamientos')
axes[0].set_ylabel('Frecuencia relativa de "Cara"')
axes[0].set_title('Convergencia de la Frecuencia Relativa (Ley de los Grandes Números)')
axes[0].legend()
axes[0].grid(True, alpha=0.3)
axes[0].set_xlim(0, max_lanzamientos)
axes[0].set_ylim(0.3, 0.7)

# Gráfico 2: Error absoluto vs número de lanzamientos (escala logarítmica)
error_absoluto = np.abs(frecuencia_relativa - prob_teorica)
axes[1].plot(n_lanzamientos, error_absoluto, color='orange', linewidth=1.5)
axes[1].set_xlabel('Número de lanzamientos')
axes[1].set_ylabel('Error absoluto |freq - prob|')
axes[1].set_title('Reducción del Error con más Lanzamientos')
axes[1].set_xscale('log')
axes[1].set_yscale('log')
axes[1].grid(True, which='both', alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\n🎯 Conclusión: La teoría frecuencial se verifica empíricamente.")
print(f"   A mayor número de repeticiones, la frecuencia relativa converge")
print(f"   a la probabilidad teórica (0.5 en este caso).")
```
