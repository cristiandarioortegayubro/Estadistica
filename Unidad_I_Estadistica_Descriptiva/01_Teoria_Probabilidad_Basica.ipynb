# Unidad I: Estadística Descriptiva - Teoría de Probabilidad Básica

## Introducción

La **probabilidad** es una rama de la matemática que estudia los fenómenos aleatorios, es decir, aquellos cuyo resultado no puede predecirse con certeza. En el ámbito de los negocios, la probabilidad es fundamental para:

* Evaluar riesgos financieros
* Predecir demanda de productos
* Tomar decisiones bajo incertidumbre
* Analizar tendencias del mercado

## Objetivos de aprendizaje

* Comprender el concepto de probabilidad desde la definición clásica y la teoría frecuencial
* Identificar y clasificar experimentos y sucesos aleatorios
* Aplicar las reglas de la suma y del producto
* Interpretar y utilizar la probabilidad condicional
* Analizar situaciones de dependencia e independencia entre sucesos

## 1. Definiciones Fundamentales

### Experimento Aleatorio

Un **experimento aleatorio** es aquel que cumple las siguientes condiciones:

1. Se puede repetir bajo las mismas condiciones
2. No se puede predecir el resultado con certeza
3. Se conocen todos los posibles resultados

**Ejemplos:**
* Lanzar una moneda
* Lanzar un dado
* Extraer una carta de una baraja
* Tiempo que tarda un cliente en ser atendido
* Número de ventas diarias de un producto

### Espacio Muestral (Ω)

Es el conjunto de todos los posibles resultados de un experimento aleatorio.

**Ejemplos:**
* Lanzar una moneda: Ω = {Cara, Cruz}
* Lanzar un dado: Ω = {1, 2, 3, 4, 5, 6}
* Extraer una carta: Ω = {52 cartas posibles}

### Suceso o Evento

Un **suceso** es cualquier subconjunto del espacio muestral.

**Tipos de sucesos:**
* **Suceso elemental:** contiene un solo resultado
* **Suceso compuesto:** contiene más de un resultado
* **Suceso seguro:** el propio espacio muestral Ω
* **Suceso imposible:** el conjunto vacío ∅

## 2. Definición Clásica de Probabilidad (Laplace)

Para experimentos donde todos los resultados son **igualmente probables**, la probabilidad de un suceso A es:

$$P(A) = \frac{\text{Número de casos favorables a A}}{\text{Número total de casos posibles}}$$

### Propiedades

* $0 ≤ P(A) ≤ 1$ para todo suceso A
* $P(Ω) = 1$ (probabilidad del suceso seguro)
* $P(∅) = 0$ (probabilidad del suceso imposible)

### Ejemplo 1: Lanzamiento de un dado

¿Cuál es la probabilidad de obtener un número par?

* Espacio muestral: Ω = {1, 2, 3, 4, 5, 6}
* Suceso A = "obtener número par" = {2, 4, 6}
* $P(A) = \frac{3}{6} = \frac{1}{2} = 0.5$

### Ejemplo 2: Selección de personal

Una empresa tiene 20 empleados: 12 mujeres y 8 hombres. Si se selecciona uno al azar para una capacitación, ¿cuál es la probabilidad de que sea mujer?

* $P(\text{Mujer}) = \frac{12}{20} = 0.6 = 60\%$

## 3. Teoría Frecuencial de Probabilidad

La **frecuencia relativa** de un suceso A en n repeticiones de un experimento es:

$$f_n(A) = \frac{\text{Número de veces que ocurre A}}{n}$$

Según la **Ley de los Grandes Números**, cuando el número de repeticiones tiende a infinito, la frecuencia relativa se estabiliza en un valor que llamamos **probabilidad**:

$$P(A) = \lim_{n \to \infty} f_n(A)$$

### Características

* Se basa en la repetición del experimento
* Es útil cuando no se pueden aplicar métodos clásicos
* Permite estimar probabilidades empíricamente
* Se utiliza en situaciones reales de negocios

### Ejemplo: Tasa de defectos

Una fábrica analiza 1000 productos y encuentra 25 defectuosos.

* Frecuencia relativa de defectos: $f_{1000} = \frac{25}{1000} = 0.025$
* Estimación de la probabilidad: $P(\text{Defecto}) ≈ 0.025 = 2.5\%$

Esta es una **estimación** de la verdadera probabilidad basada en la muestra observada.

## 4. Operaciones con Sucesos

### Sucesos Mutuamente Excluyentes

Dos sucesos A y B son **mutuamente excluyentes** (o incompatibles) si no pueden ocurrir simultáneamente:

$$A ∩ B = ∅$$

**Ejemplo:** Al lanzar un dado, los sucesos "obtener un 2" y "obtener un 5" son mutuamente excluyentes.

### Sucesos No Mutuamente Excluyentes

Dos sucesos pueden ocurrir al mismo tiempo:

$$A ∩ B ≠ ∅$$

**Ejemplo:** Al extraer una carta, los sucesos "obtener un As" y "obtener un trébol" no son mutuamente excluyentes (el As de tréboles satisface ambos).

### Suceso Complementario

El complemento de A (denotado $A^c$ o $\bar{A}$) es el conjunto de resultados que no están en A:

$$P(A^c) = 1 - P(A)$$

**Ejemplo:** Si $P(\text{Lluvia}) = 0.3$, entonces $P(\text{No lluvia}) = 0.7$

## 5. Reglas Fundamentales de Probabilidad

### Regla de la Suma (Unión de Sucesos)

Para dos sucesos A y B **cualesquiera**:

$$P(A ∪ B) = P(A) + P(B) - P(A ∩ B)$$

**Caso especial:** Si A y B son mutuamente excluyentes:

$$P(A ∪ B) = P(A) + P(B)$$

**Ejemplo:** Una empresa analiza riesgos. La probabilidad de pérdidas por:
* Cambio de regulación: P(A) = 0.15
* Problemas operativos: P(B) = 0.20
* Ambos simultáneamente: P(A ∩ B) = 0.05

¿Probabilidad de al menos uno de estos problemas?

$$P(A ∪ B) = 0.15 + 0.20 - 0.05 = 0.30$$

### Regla del Producto (Intersección de Sucesos)

Para dos sucesos A y B:

$$P(A ∩ B) = P(A) \cdot P(B|A)$$

Donde $P(B|A)$ es la probabilidad de B **dado que** ocurrió A (probabilidad condicional).

## 6. Probabilidad Condicional

La **probabilidad condicional** de B dado A (cuando $P(A) > 0$) es:

$$P(B|A) = \frac{P(A ∩ B)}{P(A)}$$

**Interpretación:** Es la probabilidad de que ocurra B, sabiendo que ya ocurrió A.

### Ejemplo: Marketing dirigido

Una empresa observa que:
* 60% de los clientes son mujeres: P(M) = 0.6
* 15% de todos los clientes compran el producto premium: P(P) = 0.15
* 12% son mujeres que compran premium: P(M ∩ P) = 0.12

¿Cuál es la probabilidad de que una cliente mujer compre el producto premium?

$$P(P|M) = \frac{P(M ∩ P)}{P(M)} = \frac{0.12}{0.6} = 0.20$$

Es decir, el 20% de las mujeres compran premium, aunque solo el 15% de todos los clientes lo hacen.

### Fórmula del Producto (reformulada)

Despejando de la fórmula de probabilidad condicional:

$$P(A ∩ B) = P(A) \cdot P(B|A) = P(B) \cdot P(A|B)$$

## 7. Sucesos Independientes

Dos sucesos A y B son **independientes** si la ocurrencia de uno no afecta la probabilidad del otro:

$$P(B|A) = P(B)$$

**Equivalentemente:**

$$P(A ∩ B) = P(A) \cdot P(B)$$

### Diferencia importante

⚠️ **Independencia ≠ Mutua exclusión**

* **Mutuamente excluyentes:** No pueden ocurrir juntos ($A ∩ B = ∅$)
* **Independientes:** La ocurrencia de uno no afecta al otro

De hecho, si dos sucesos son mutuamente excluyentes con probabilidad positiva, entonces **NO son independientes** (porque si ocurre uno, el otro definitivamente no puede ocurrir).

### Ejemplo 1: Lanzamientos de moneda

Lanzar una moneda dos veces:
* A = "Cara en el primer lanzamiento"
* B = "Cara en el segundo lanzamiento"

Estos sucesos son independientes: el resultado del primer lanzamiento no afecta al segundo.

$$P(A ∩ B) = P(A) \cdot P(B) = 0.5 \times 0.5 = 0.25$$

### Ejemplo 2: Control de calidad

Una empresa produce componentes con 2% de defectos. Si se seleccionan dos componentes al azar (con reemplazo):

* A = "Primer componente defectuoso": P(A) = 0.02
* B = "Segundo componente defectuoso": P(B) = 0.02

¿Probabilidad de que ambos sean defectuosos?

$$P(A ∩ B) = P(A) \cdot P(B) = 0.02 \times 0.02 = 0.0004$$

Solo 0.04% de probabilidad.

## 8. Resumen de Fórmulas Clave

### Propiedades básicas
* $0 ≤ P(A) ≤ 1$
* $P(Ω) = 1$
* $P(A^c) = 1 - P(A)$

### Reglas de operación
* **Regla de la suma:** $P(A ∪ B) = P(A) + P(B) - P(A ∩ B)$
* **Si son mutuamente excluyentes:** $P(A ∪ B) = P(A) + P(B)$

### Probabilidad condicional
* **Definición:** $P(B|A) = \frac{P(A ∩ B)}{P(A)}$
* **Regla del producto:** $P(A ∩ B) = P(A) \cdot P(B|A)$

### Independencia
* **Definición:** $P(B|A) = P(B)$
* **Equivalente:** $P(A ∩ B) = P(A) \cdot P(B)$
