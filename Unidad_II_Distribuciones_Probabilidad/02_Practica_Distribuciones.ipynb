## Ejercicio 1: Distribución Bernoulli

### Problema
Un cliente tiene 30% de probabilidad de comprar un producto al visitar una tienda online.

**Preguntas:**
1. ¿Cuál es la probabilidad de que compre?
2. ¿Cuál es la esperanza y varianza?
3. Simular 1000 visitas y verificar las propiedades.

```python
# Parámetro
p = 0.30

# Crear distribución Bernoulli
X = stats.bernoulli(p)

print("=" * 60)
print("DISTRIBUCIÓN BERNOULLI: Cliente compra producto")
print("=" * 60)

# 1. Probabilidades
print(f"\n1. Probabilidades:")
print(f"   P(X = 0) [No compra] = {X.pmf(0):.4f}")
print(f"   P(X = 1) [Compra]    = {X.pmf(1):.4f}")

# 2. Parámetros teóricos
print(f"\n2. Parámetros teóricos:")
print(f"   E(X) = {X.mean():.4f}")
print(f"   Var(X) = {X.var():.4f}")
print(f"   σ(X) = {X.std():.4f}")

# 3. Simulación
np.random.seed(42)
simulaciones = X.rvs(size=1000)

print(f"\n3. Simulación (1000 visitas):")
print(f"   Compras realizadas: {simulaciones.sum()}")
print(f"   Proporción de compras: {simulaciones.mean():.4f}")
print(f"   Varianza muestral: {simulaciones.var():.4f}")

# Visualización
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Gráfico 1: Función de probabilidad
valores = [0, 1]
probs = [X.pmf(x) for x in valores]
ax1.bar(valores, probs, color=['red', 'green'], alpha=0.7, edgecolor='black')
ax1.set_xlabel('Resultado (0=No compra, 1=Compra)')
ax1.set_ylabel('Probabilidad')
ax1.set_title('Función de Probabilidad Bernoulli(0.30)')
ax1.set_xticks([0, 1])
ax1.grid(axis='y', alpha=0.3)

# Gráfico 2: Histograma de simulación
ax2.hist(simulaciones, bins=[-0.5, 0.5, 1.5], density=True, 
         color='skyblue', alpha=0.7, edgecolor='black', label='Simulación')
ax2.bar(valores, probs, color='red', alpha=0.3, width=0.3, label='Teórico')
ax2.set_xlabel('Resultado')
ax2.set_ylabel('Frecuencia relativa')
ax2.set_title('Simulación vs Teórico')
ax2.set_xticks([0, 1])
ax2.legend()
ax2.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

## Ejercicio 2: Distribución Binomial

### Problema
Una empresa de marketing envía correos a 50 clientes. La tasa de conversión histórica es del 15%.

**Preguntas:**
1. ¿Cuál es la probabilidad de obtener exactamente 10 conversiones?
2. ¿Cuál es la probabilidad de obtener al menos 5 conversiones?
3. ¿Cuántas conversiones se esperan?
4. Graficar la distribución.

```python
# Parámetros
n = 50  # número de ensayos
p = 0.15  # probabilidad de éxito

# Crear distribución Binomial
X = stats.binom(n, p)

print("=" * 60)
print("DISTRIBUCIÓN BINOMIAL: Conversiones de correos")
print("=" * 60)
print(f"n = {n} correos, p = {p} tasa de conversión\n")

# 1. Probabilidad exacta
prob_exacta = X.pmf(10)
print(f"1. P(X = 10) = {prob_exacta:.4f} ({prob_exacta*100:.2f}%)")

# 2. Probabilidad acumulada
prob_al_menos_5 = 1 - X.cdf(4)  # P(X >= 5) = 1 - P(X <= 4)
print(f"\n2. P(X ≥ 5) = {prob_al_menos_5:.4f} ({prob_al_menos_5*100:.2f}%)")

# Método alternativo
prob_al_menos_5_alt = X.sf(4)  # Survival function: 1 - CDF(4)
print(f"   Verificación: {prob_al_menos_5_alt:.4f}")

# 3. Esperanza y varianza
print(f"\n3. Parámetros:")
print(f"   E(X) = np = {X.mean():.2f} conversiones esperadas")
print(f"   Var(X) = np(1-p) = {X.var():.2f}")
print(f"   σ(X) = {X.std():.2f}")

# 4. Probabilidades de los valores más comunes
print(f"\n4. Probabilidades de valores comunes:")
for k in [5, 7, 10, 12, 15]:
    print(f"   P(X = {k:2d}) = {X.pmf(k):.4f}")

# Visualización
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Gráfico 1: Función de probabilidad
valores = np.arange(0, n+1)
probs = X.pmf(valores)
ax1.bar(valores, probs, color='steelblue', alpha=0.7, edgecolor='black')
ax1.axvline(X.mean(), color='red', linestyle='--', linewidth=2, label=f'E(X) = {X.mean():.1f}')
ax1.axvline(10, color='green', linestyle='--', linewidth=2, label='X = 10')
ax1.set_xlabel('Número de conversiones')
ax1.set_ylabel('Probabilidad')
ax1.set_title(f'Distribución Binomial(n={n}, p={p})')
ax1.legend()
ax1.grid(axis='y', alpha=0.3)

# Gráfico 2: Función de distribución acumulada
ax2.step(valores, X.cdf(valores), where='post', color='darkblue', linewidth=2)
ax2.fill_between(valores, 0, X.cdf(valores), step='post', alpha=0.3)
ax2.axhline(0.5, color='red', linestyle='--', alpha=0.5, label='Mediana')
ax2.set_xlabel('Número de conversiones')
ax2.set_ylabel('Probabilidad acumulada')
ax2.set_title('Función de Distribución Acumulada (CDF)')
ax2.legend()
ax2.grid(alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\n✓ La distribución es aproximadamente simétrica alrededor de E(X) = {X.mean():.1f}")
```

## Ejercicio 3: Distribución de Poisson

### Problema
Un call center recibe en promedio 20 llamadas por hora.

**Preguntas:**
1. ¿Cuál es la probabilidad de recibir exactamente 25 llamadas en una hora?
2. ¿Cuál es la probabilidad de recibir menos de 15 llamadas?
3. ¿Cuál es la probabilidad de recibir entre 18 y 22 llamadas?
4. Simular 100 horas de operación.

```python
# Parámetro
lambda_rate = 20  # tasa promedio por hora

# Crear distribución Poisson
X = stats.poisson(lambda_rate)

print("=" * 60)
print("DISTRIBUCIÓN DE POISSON: Llamadas al call center")
print("=" * 60)
print(f"λ = {lambda_rate} llamadas/hora\n")

# 1. Probabilidad exacta
prob_25 = X.pmf(25)
print(f"1. P(X = 25) = {prob_25:.4f} ({prob_25*100:.2f}%)")

# 2. Probabilidad menor que
prob_menos_15 = X.cdf(14)  # P(X < 15) = P(X <= 14)
print(f"\n2. P(X < 15) = {prob_menos_15:.4f} ({prob_menos_15*100:.2f}%)")

# 3. Probabilidad entre valores
prob_entre = X.cdf(22) - X.cdf(17)  # P(18 ≤ X ≤ 22)
print(f"\n3. P(18 ≤ X ≤ 22) = {prob_entre:.4f} ({prob_entre*100:.2f}%)")

# Parámetros
print(f"\n4. Parámetros:")
print(f"   E(X) = λ = {X.mean():.2f}")
print(f"   Var(X) = λ = {X.var():.2f}")
print(f"   σ(X) = √λ = {X.std():.2f}")

# 5. Simulación
np.random.seed(42)
simulacion = X.rvs(size=100)

print(f"\n5. Simulación (100 horas):")
print(f"   Media de llamadas: {simulacion.mean():.2f}")
print(f"   Varianza: {simulacion.var():.2f}")
print(f"   Mínimo: {simulacion.min()} llamadas")
print(f"   Máximo: {simulacion.max()} llamadas")

# Visualización
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Gráfico 1: Función de probabilidad
valores = np.arange(0, 45)
probs = X.pmf(valores)
axes[0, 0].bar(valores, probs, color='coral', alpha=0.7, edgecolor='black')
axes[0, 0].axvline(lambda_rate, color='red', linestyle='--', linewidth=2, label=f'λ = {lambda_rate}')
axes[0, 0].set_xlabel('Número de llamadas')
axes[0, 0].set_ylabel('Probabilidad')
axes[0, 0].set_title(f'Distribución de Poisson(λ={lambda_rate})')
axes[0, 0].legend()
axes[0, 0].grid(axis='y', alpha=0.3)

# Gráfico 2: CDF
axes[0, 1].step(valores, X.cdf(valores), where='post', color='darkred', linewidth=2)
axes[0, 1].fill_between(valores, 0, X.cdf(valores), step='post', alpha=0.3, color='coral')
axes[0, 1].set_xlabel('Número de llamadas')
axes[0, 1].set_ylabel('Probabilidad acumulada')
axes[0, 1].set_title('Función de Distribución Acumulada')
axes[0, 1].grid(alpha=0.3)

# Gráfico 3: Histograma de simulación
axes[1, 0].hist(simulacion, bins=20, density=True, color='lightcoral', 
                alpha=0.7, edgecolor='black', label='Simulación (100 horas)')
valores_plot = np.arange(0, 40)
axes[1, 0].plot(valores_plot, X.pmf(valores_plot), 'o-', color='red', 
                markersize=4, label='Teórico', alpha=0.6)
axes[1, 0].set_xlabel('Número de llamadas')
axes[1, 0].set_ylabel('Densidad')
axes[1, 0].set_title('Simulación vs Teórico')
axes[1, 0].legend()
axes[1, 0].grid(alpha=0.3)

# Gráfico 4: Boxplot de simulación
axes[1, 1].boxplot(simulacion, vert=True)
axes[1, 1].axhline(lambda_rate, color='red', linestyle='--', linewidth=2, label=f'λ = {lambda_rate}')
axes[1, 1].set_ylabel('Número de llamadas')
axes[1, 1].set_title('Boxplot de 100 horas simuladas')
axes[1, 1].legend()
axes[1, 1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\n✓ Característica: E(X) = Var(X) = λ = {lambda_rate}")
```

## Ejercicio 4: Distribución Normal

### Problema
Las ventas diarias de una tienda siguen una distribución normal con media $5,000 y desviación estándar $800.

**Preguntas:**
1. ¿Cuál es la probabilidad de que las ventas estén entre $4,500 y $6,000?
2. ¿Cuál es la probabilidad de que las ventas superen $6,500?
3. ¿Entre qué valores están el 95% central de las ventas?
4. Aplicar la regla empírica (68-95-99.7).

```python
# Parámetros
mu = 5000  # media
sigma = 800  # desviación estándar

# Crear distribución Normal
X = stats.norm(loc=mu, scale=sigma)

print("=" * 60)
print("DISTRIBUCIÓN NORMAL: Ventas diarias")
print("=" * 60)
print(f"μ = ${mu:,}, σ = ${sigma:,}\n")

# 1. Probabilidad entre valores
prob_entre = X.cdf(6000) - X.cdf(4500)
print(f"1. P(4500 ≤ X ≤ 6000) = {prob_entre:.4f} ({prob_entre*100:.2f}%)")

# Cálculo con Z-scores
z1 = (4500 - mu) / sigma
z2 = (6000 - mu) / sigma
print(f"   Z-scores: z1 = {z1:.2f}, z2 = {z2:.2f}")

# 2. Probabilidad mayor que
prob_mayor = 1 - X.cdf(6500)
print(f"\n2. P(X > 6500) = {prob_mayor:.4f} ({prob_mayor*100:.2f}%)")
z_6500 = (6500 - mu) / sigma
print(f"   Z-score: z = {z_6500:.2f}")

# 3. Intervalo del 95% central
# Queremos el intervalo [a, b] tal que P(a ≤ X ≤ b) = 0.95
# con igual probabilidad en las colas: P(X < a) = 0.025 y P(X > b) = 0.025
limite_inferior = X.ppf(0.025)  # percentil 2.5%
limite_superior = X.ppf(0.975)  # percentil 97.5%
print(f"\n3. Intervalo del 95% central:")
print(f"   [${limite_inferior:.2f}, ${limite_superior:.2f}]")
print(f"   También: μ ± 1.96σ = [{mu - 1.96*sigma:.2f}, {mu + 1.96*sigma:.2f}]")

# 4. Regla empírica
print(f"\n4. Regla empírica (68-95-99.7):")
print(f"   68% de las ventas están en μ ± 1σ:")
print(f"      [${mu-sigma:.2f}, ${mu+sigma:.2f}]")
prob_1sigma = X.cdf(mu+sigma) - X.cdf(mu-sigma)
print(f"      Probabilidad real: {prob_1sigma:.4f} ({prob_1sigma*100:.2f}%)")

print(f"\n   95% de las ventas están en μ ± 2σ:")
print(f"      [${mu-2*sigma:.2f}, ${mu+2*sigma:.2f}]")
prob_2sigma = X.cdf(mu+2*sigma) - X.cdf(mu-2*sigma)
print(f"      Probabilidad real: {prob_2sigma:.4f} ({prob_2sigma*100:.2f}%)")

print(f"\n   99.7% de las ventas están en μ ± 3σ:")
print(f"      [${mu-3*sigma:.2f}, ${mu+3*sigma:.2f}]")
prob_3sigma = X.cdf(mu+3*sigma) - X.cdf(mu-3*sigma)
print(f"      Probabilidad real: {prob_3sigma:.4f} ({prob_3sigma*100:.2f}%)")

# Visualización
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Rango de valores para graficar
x_vals = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y_vals = X.pdf(x_vals)

# Gráfico 1: Densidad con áreas sombreadas
axes[0, 0].plot(x_vals, y_vals, 'b-', linewidth=2, label='Densidad')
axes[0, 0].fill_between(x_vals, 0, y_vals, where=(x_vals >= 4500) & (x_vals <= 6000),
                         color='lightblue', alpha=0.5, label='P(4500 ≤ X ≤ 6000)')
axes[0, 0].axvline(mu, color='red', linestyle='--', linewidth=2, label=f'μ = {mu}')
axes[0, 0].set_xlabel('Ventas ($)')
axes[0, 0].set_ylabel('Densidad')
axes[0, 0].set_title(f'Distribución Normal(μ={mu}, σ={sigma})')
axes[0, 0].legend()
axes[0, 0].grid(alpha=0.3)

# Gráfico 2: Regla empírica
axes[0, 1].plot(x_vals, y_vals, 'b-', linewidth=2)
for k in [1, 2, 3]:
    axes[0, 1].axvspan(mu - k*sigma, mu + k*sigma, alpha=0.1*k, 
                       color=['green', 'yellow', 'red'][k-1])
    axes[0, 1].axvline(mu - k*sigma, color='gray', linestyle='--', alpha=0.5)
    axes[0, 1].axvline(mu + k*sigma, color='gray', linestyle='--', alpha=0.5)
axes[0, 1].axvline(mu, color='red', linestyle='--', linewidth=2)
axes[0, 1].set_xlabel('Ventas ($)')
axes[0, 1].set_ylabel('Densidad')
axes[0, 1].set_title('Regla Empírica: 68%-95%-99.7%')
axes[0, 1].text(mu, max(y_vals)*0.9, '68%', ha='center', fontsize=12)
axes[0, 1].text(mu, max(y_vals)*0.7, '95%', ha='center', fontsize=12)
axes[0, 1].text(mu, max(y_vals)*0.5, '99.7%', ha='center', fontsize=12)
axes[0, 1].grid(alpha=0.3)

# Gráfico 3: CDF
axes[1, 0].plot(x_vals, X.cdf(x_vals), 'darkblue', linewidth=2)
axes[1, 0].axhline(0.5, color='red', linestyle='--', alpha=0.5, label='Mediana')
axes[1, 0].axvline(mu, color='red', linestyle='--', alpha=0.5)
axes[1, 0].set_xlabel('Ventas ($)')
axes[1, 0].set_ylabel('Probabilidad acumulada')
axes[1, 0].set_title('Función de Distribución Acumulada')
axes[1, 0].legend()
axes[1, 0].grid(alpha=0.3)

# Gráfico 4: Simulación
np.random.seed(42)
simulacion = X.rvs(size=1000)
axes[1, 1].hist(simulacion, bins=30, density=True, color='lightblue', 
                alpha=0.7, edgecolor='black', label='Simulación (1000 días)')
axes[1, 1].plot(x_vals, y_vals, 'r-', linewidth=2, label='Teórico')
axes[1, 1].axvline(simulacion.mean(), color='green', linestyle='--', 
                   linewidth=2, label=f'Media simulación = ${simulacion.mean():.0f}')
axes[1, 1].set_xlabel('Ventas ($)')
axes[1, 1].set_ylabel('Densidad')
axes[1, 1].set_title('Simulación vs Teórico')
axes[1, 1].legend()
axes[1, 1].grid(alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\n✓ La distribución normal es simétrica alrededor de μ = ${mu:,}")
```

## Ejercicio 5: Estandarización (Z-scores)

### Problema
Los tiempos de entrega de un servicio siguen N(25, 4²) minutos.

**Preguntas:**
1. Estandarizar tiempos de 20, 25 y 30 minutos
2. ¿Qué tiempo corresponde a un Z-score de 1.5?
3. Comparar dos distribuciones usando estandarización

```python
# Parámetros
mu = 25
sigma = 4

X = stats.norm(loc=mu, scale=sigma)
Z = stats.norm(loc=0, scale=1)  # Normal estándar

print("=" * 60)
print("ESTANDARIZACIÓN: Tiempos de entrega")
print("=" * 60)
print(f"X ~ N(μ={mu}, σ²={sigma**2})\n")

# 1. Calcular Z-scores
tiempos = [20, 25, 30]
print("1. Z-scores para diferentes tiempos:")
for t in tiempos:
    z = (t - mu) / sigma
    prob = X.cdf(t)
    print(f"   Tiempo {t:2d} min: Z = {z:+.2f}, P(X ≤ {t}) = {prob:.4f}")
    
    # Interpretación
    if z == 0:
        print(f"              → Exactamente en la media")
    elif z > 0:
        print(f"              → {abs(z):.1f} desviaciones estándar por encima de la media")
    else:
        print(f"              → {abs(z):.1f} desviaciones estándar por debajo de la media")
    print()

# 2. Valor correspondiente a Z-score
z_objetivo = 1.5
x_objetivo = mu + z_objetivo * sigma
print(f"2. ¿Qué tiempo corresponde a Z = {z_objetivo}?")
print(f"   x = μ + Z·σ = {mu} + {z_objetivo}·{sigma} = {x_objetivo:.2f} minutos")
print(f"   P(X ≤ {x_objetivo:.2f}) = {X.cdf(x_objetivo):.4f}")

# 3. Comparación de distribuciones
print(f"\n3. Comparación usando estandarización:")
print(f"\n   Servicio A: μ_A = 25 min, σ_A = 4 min")
print(f"   Servicio B: μ_B = 30 min, σ_B = 5 min")
print(f"\n   Si A tardó 29 min y B tardó 35 min, ¿cuál fue relativamente más rápido?")

# Servicio A
mu_A, sigma_A = 25, 4
tiempo_A = 29
z_A = (tiempo_A - mu_A) / sigma_A
print(f"\n   Servicio A: Z_A = ({tiempo_A} - {mu_A}) / {sigma_A} = {z_A:.2f}")

# Servicio B
mu_B, sigma_B = 30, 5
tiempo_B = 35
z_B = (tiempo_B - mu_B) / sigma_B
print(f"   Servicio B: Z_B = ({tiempo_B} - {mu_B}) / {sigma_B} = {z_B:.2f}")

print(f"\n   Conclusión: Servicio {'A' if z_A < z_B else 'B'} fue relativamente más rápido")
print(f"   (menor Z-score significa más cerca de su propia media)")

# Visualización
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Gráfico 1: Distribución original y estandarizada
x_vals = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y_vals = X.pdf(x_vals)

z_vals = np.linspace(-4, 4, 1000)
z_y_vals = Z.pdf(z_vals)

ax1.plot(x_vals, y_vals, 'b-', linewidth=2, label=f'X ~ N({mu}, {sigma}²)')
ax1_twin = ax1.twiny()
ax1_twin.plot(z_vals, z_y_vals, 'r--', linewidth=2, label='Z ~ N(0, 1)')

for t in tiempos:
    z = (t - mu) / sigma
    ax1.axvline(t, color='gray', linestyle='--', alpha=0.5)
    ax1.text(t, max(y_vals)*1.1, f'{t}\n(Z={z:+.1f})', ha='center', fontsize=9)

ax1.set_xlabel('Tiempo (minutos)')
ax1_twin.set_xlabel('Z-score')
ax1.set_ylabel('Densidad')
ax1.set_title('Distribución Original vs Estandarizada')
ax1.legend(loc='upper left')
ax1_twin.legend(loc='upper right')
ax1.grid(alpha=0.3)

# Gráfico 2: Normal estándar con áreas
ax2.plot(z_vals, z_y_vals, 'r-', linewidth=2)
for z in [-3, -2, -1, 0, 1, 2, 3]:
    ax2.axvline(z, color='gray', linestyle='--', alpha=0.3)
    ax2.text(z, -0.01, f'{z}', ha='center', fontsize=10)

# Sombrear áreas
ax2.fill_between(z_vals, 0, z_y_vals, where=(z_vals >= -1) & (z_vals <= 1),
                 color='green', alpha=0.2, label='68% (±1σ)')
ax2.fill_between(z_vals, 0, z_y_vals, where=((z_vals >= -2) & (z_vals < -1)) | ((z_vals > 1) & (z_vals <= 2)),
                 color='yellow', alpha=0.2, label='27% (entre ±1σ y ±2σ)')
ax2.fill_between(z_vals, 0, z_y_vals, where=((z_vals >= -3) & (z_vals < -2)) | ((z_vals > 2) & (z_vals <= 3)),
                 color='red', alpha=0.2, label='4.3% (entre ±2σ y ±3σ)')

ax2.set_xlabel('Z-score')
ax2.set_ylabel('Densidad')
ax2.set_title('Distribución Normal Estándar Z ~ N(0, 1)')
ax2.legend()
ax2.grid(alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\n✓ La estandarización permite comparar valores de diferentes distribuciones")
```

## Ejercicio 6: Verificación del Teorema Central del Límite

### Objetivo
Demostrar que la distribución de medias muestrales tiende a ser normal, independientemente de la distribución de la población.

```python
print("=" * 60)
print("TEOREMA CENTRAL DEL LÍMITE: Simulación")
print("=" * 60)

# Crear una población con distribución NO NORMAL (exponencial)
np.random.seed(42)
poblacion = np.random.exponential(scale=2, size=10000)

print(f"\nPoblación original (Exponencial):")
print(f"  Media: {poblacion.mean():.3f}")
print(f"  Desv. Est.: {poblacion.std():.3f}")
print(f"  Tamaño: {len(poblacion)}")

# Tomar muchas muestras de diferentes tamaños
tamaños_muestra = [5, 10, 30, 50]
num_muestras = 1000

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

for idx, n in enumerate(tamaños_muestra):
    # Tomar num_muestras muestras de tamaño n
    medias_muestrales = []
    for _ in range(num_muestras):
        muestra = np.random.choice(poblacion, size=n)
        medias_muestrales.append(muestra.mean())
    
    medias_muestrales = np.array(medias_muestrales)
    
    # Calcular estadísticas
    media_de_medias = medias_muestrales.mean()
    std_de_medias = medias_muestrales.std()
    
    # Error estándar teórico
    error_estandar_teorico = poblacion.std() / np.sqrt(n)
    
    print(f"\nMuestras de tamaño n = {n}:")
    print(f"  Media de las medias: {media_de_medias:.3f}")
    print(f"  Desv. Est. de las medias: {std_de_medias:.3f}")
    print(f"  Error estándar teórico (σ/√n): {error_estandar_teorico:.3f}")
    
    # Graficar
    ax = axes[idx]
    
    # Histograma de medias muestrales
    ax.hist(medias_muestrales, bins=30, density=True, 
            color='skyblue', alpha=0.7, edgecolor='black', label='Medias muestrales')
    
    # Curva normal teórica
    x_vals = np.linspace(medias_muestrales.min(), medias_muestrales.max(), 100)
    normal_teorica = stats.norm.pdf(x_vals, loc=poblacion.mean(), scale=error_estandar_teorico)
    ax.plot(x_vals, normal_teorica, 'r-', linewidth=2, label='Normal teórica')
    
    ax.axvline(poblacion.mean(), color='green', linestyle='--', 
               linewidth=2, label=f'μ = {poblacion.mean():.2f}')
    
    ax.set_xlabel('Media muestral')
    ax.set_ylabel('Densidad')
    ax.set_title(f'n = {n} (Error Est. = {error_estandar_teorico:.3f})')
    ax.legend()
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("\n" + "=" * 60)
print("CONCLUSIONES DEL TCL:")
print("=" * 60)
print("\n1. A medida que n aumenta, la distribución de medias se aproxima")
print("   más a una distribución normal")
print("\n2. La media de las medias muestrales ≈ media poblacional")
print("\n3. El error estándar disminuye con √n")
print("\n4. ¡Funciona incluso con población NO NORMAL!")

# Comparar distribución original vs distribución de medias
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Población original
ax1.hist(poblacion, bins=50, density=True, color='lightcoral', 
         alpha=0.7, edgecolor='black')
ax1.set_xlabel('Valor')
ax1.set_ylabel('Densidad')
ax1.set_title('Población Original (Exponencial)\nClaramente NO Normal')
ax1.grid(alpha=0.3)

# Distribución de medias (n=30)
n = 30
medias_finales = []
for _ in range(1000):
    muestra = np.random.choice(poblacion, size=n)
    medias_finales.append(muestra.mean())

ax2.hist(medias_finales, bins=30, density=True, color='lightgreen', 
         alpha=0.7, edgecolor='black', label='Medias muestrales')

# Curva normal
x_vals = np.linspace(min(medias_finales), max(medias_finales), 100)
y_vals = stats.norm.pdf(x_vals, loc=poblacion.mean(), 
                        scale=poblacion.std()/np.sqrt(n))
ax2.plot(x_vals, y_vals, 'r-', linewidth=2, label='Normal')

ax2.set_xlabel('Media muestral')
ax2.set_ylabel('Densidad')
ax2.set_title(f'Distribución de Medias (n={n})\n¡Aproximadamente Normal!')
ax2.legend()
ax2.grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("\n✓ Teorema Central del Límite verificado experimentalmente")
```

## Resumen y Conclusiones

### Distribuciones Cubiertas

| Distribución | Tipo | Uso Principal | scipy.stats |
|--------------|------|---------------|-------------|
| Bernoulli | Discreta | Un ensayo éxito/fracaso | `bernoulli(p)` |
| Binomial | Discreta | n ensayos independientes | `binom(n, p)` |
| Poisson | Discreta | Eventos en intervalo | `poisson(λ)` |
| Normal | Continua | Variables simétricas | `norm(μ, σ)` |

### Métodos Principales en scipy.stats

* **`.pmf(k)`**: Función de probabilidad (discretas) - P(X = k)
* **`.pdf(x)`**: Función de densidad (continuas) - f(x)
* **`.cdf(x)`**: Función acumulada - P(X ≤ x)
* **`.sf(x)`**: Función de supervivencia - P(X > x) = 1 - CDF(x)
* **`.ppf(q)`**: Percentil (inversa de CDF) - valor x tal que P(X ≤ x) = q
* **`.rvs(size)`**: Generar muestras aleatorias
* **`.mean()`**: Esperanza E(X)
* **`.var()`**: Varianza Var(X)
* **`.std()`**: Desviación estándar σ(X)

### Conceptos Clave

1. **Variables Aleatorias**: Funciones que asignan valores numéricos a resultados aleatorios
2. **Esperanza**: Valor promedio o centro de la distribución
3. **Varianza**: Medida de dispersión
4. **Estandarización**: Transformar a Z ~ N(0,1) para facilitar cálculos
5. **TCL**: Las medias muestrales tienden a ser normales para n grande
